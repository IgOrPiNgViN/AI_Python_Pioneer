## Модуль 1. Старт и инструменты (уроки 1–2)

**Урок 2. Промпт-инженерия.** Запрос к ИИ-помощнику: **роль → контекст → задача → формат → ограничения**. Хороший ответ — часто со 2–3 попытки. ИИ — инструмент, не автопилот: сначала сам, потом вопрос, потом проверка. Личные данные в чат не отправляем.

````python
# Уроки 5–8 в одной ячейке: конвейер «прочитать → почистить → посчитать»
answers = [" Да ", "нет", "ДА!", "да", "Нет."]

def normalize(text):
    return text.strip().lower().replace("!", "").replace(".", "")

counts = {}
for a in answers:
    word = normalize(a)
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1
print(counts)
````

**Уроки 14–16.** Сквозная практика на новых данных и резервные тренажёры.

````python
import pandas as pd

df = pd.DataFrame({
    "кружок": ["Python", "роботы ", "Python", "Игры", "Роботы", None],
    "часы": [8, 3, None, 5, 7, 2],
    "балл": [85, 42, 91, 58, 74, 35],
})
df["кружок"] = df["кружок"].str.strip().str.capitalize().fillna("Неизвестно")
df["часы"] = df["часы"].fillna(df["часы"].median())
print(df.groupby("кружок")["балл"].agg(["mean", "count"]))
print("Корреляция часов и балла:", round(df["часы"].corr(df["балл"]), 2))
````

## Модуль 4. Машинное обучение (уроки 17–25)

**Урок 17. Постановка задачи.** Объект — строка, признаки **X**, цель **y**. Классификация — цель-класс, регрессия — цель-число. **Утечка** — признак, неизвестный в момент предсказания. `id` и имена — не признаки.

**Урок 19. Первая модель и train / test.** Любая модель sklearn: создать → `fit` → `predict`. Дерево решений — правила `if`, найденные по примерам. Качество проверяем **только на test**.

**Урок 20. kNN.** Ответ как у большинства из k ближайших соседей. Нужен **масштаб** признаков (делим на максимум **по train**). Категории → one-hot (`pd.get_dummies`).

**Урок 21. Метрики классификации.** Сначала **baseline** (`DummyClassifier`). Матрица ошибок: TP, FP, FN, TN. **Recall** — доля найденных среди искомых, **precision** — доля верных тревог. Главную метрику выбирают по цене ошибки; порог — решение людей.

**Урок 22. Регрессия.** Линейная регрессия: `y = a · x + b`. **MAE** — средняя ошибка в единицах задачи. Baseline — среднее. За пределами данных модели верить нельзя.

**Урок 23. Переобучение.** Train намного лучше test — переобучение; оба плохие — недообучение. Настройки выбирают по **validation**, test открывают один раз. Из равных моделей — более простая.

**Урок 24. Этика.** Смещение из-за исторических данных, нерепрезентативных выборок, признаков-посредников. Считай метрики **по группам**. Персональные данные — с согласием и обезличенно. Решения о людях принимает человек.

**Уроки 18 и 25** — самостоятельные работы №2 (данные) и №3 (полный цикл ML).

````python
import random
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score

random.seed(0)
rows = []
for i in range(200):
    hours = random.randint(1, 14)
    score = 20 + 5 * hours + random.gauss(0, 10)
    rows.append({"часы": hours, "сон": round(random.uniform(6, 9), 1), "сдал": score >= 60})
data = pd.DataFrame(rows)

X = data[["часы", "сон"]]
y = data["сдал"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
model = DecisionTreeClassifier(max_depth=2, random_state=0).fit(X_train, y_train)
pred = model.predict(X_test)

print("baseline:", round(accuracy_score(y_test, baseline.predict(X_test)), 2))
print("модель на test:", round(accuracy_score(y_test, pred), 2),
      "| на train:", round(accuracy_score(y_train, model.predict(X_train)), 2))
print(confusion_matrix(y_test, pred))
print("recall «не сдал»:", round(recall_score(y_test, pred, pos_label=False), 2))
````

## Главные правила инженера ИИ

1. Сначала задача и данные — потом модель.
2. Каждый признак проверяй вопросом «известен ли он в момент предсказания?».
3. Сравнивай с baseline; качество — только на данных, которых модель не видела.
4. Одной accuracy мало: смотри, **какие** ошибки делает модель и **для кого**.
5. Называй ограничения конкретно: на каких данных обучена, где нельзя применять, кому может навредить.
6. ИИ-помощник — инструмент: проверяй каждый ответ и отвечай за результат сам.
