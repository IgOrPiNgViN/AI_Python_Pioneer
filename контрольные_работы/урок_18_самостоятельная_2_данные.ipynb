## Данные

Таблица учеников кружков: часы занятий в неделю, посещаемость (%), число выполненных проектов, итоговый балл, сдал ли итоговую работу. В данные специально добавлены ошибки, как в настоящей выгрузке.

Запусти ячейку — она создаст файл `sr2_students.csv`. Код ячейки читать не обязательно (хотя всё в нём тебе знакомо).

````python
import random
import pandas as pd

random.seed(2027)
rows = []
for i in range(45):
    hours = random.randint(0, 15)
    attendance = random.randint(30, 100)
    projects = random.randint(0, 5)
    score = round(10 + 3.5 * hours + 0.25 * attendance + 4 * projects + random.gauss(0, 8))
    rows.append({
        "id": i + 1,
        "кружок": random.choice(["Python", "Роботы", "Игры", "Дизайн"]),
        "часы": hours,
        "посещаемость": attendance,
        "проектов": projects,
        "балл": min(100, max(0, score)),
    })

# Ошибки, как в настоящей выгрузке
for k in [3, 11, 20, 33]:
    rows[k]["посещаемость"] = None
for k in [7, 25, 40]:
    rows[k]["кружок"] = None
rows[5]["кружок"] = "python"
rows[14]["кружок"] = " Роботы"
rows[30]["кружок"] = "ИГРЫ"
rows[18]["часы"] = 99
rows.append(rows[2])

raw = pd.DataFrame(rows)
raw["сдал"] = raw["балл"] >= 60
raw.to_csv("sr2_students.csv", index=False)
print("Файл sr2_students.csv создан")
````

Разумные решения: `strip` + `capitalize` для кружков; удалить дубликат; строку с 99 часами удалить (или заменить на пропуск — с обоснованием); пропуски в посещаемости — медиана (+ флаг); пропуски в кружке — «Неизвестно».

```python
df = pd.read_csv("sr2_students.csv")
clean = df.copy()
print(clean.isna().sum(), clean.duplicated().sum())
clean["кружок"] = clean["кружок"].str.strip().str.capitalize().fillna("Неизвестно")
clean = clean.drop_duplicates()
clean = clean[clean["часы"] <= 15]
clean["посещаемость"] = clean["посещаемость"].fillna(clean["посещаемость"].median())
clean = clean.reset_index(drop=True)
print(clean.shape)

print(clean["кружок"].value_counts())
print(clean.groupby("кружок")["сдал"].agg(["mean", "count"]).round(2))
print(clean.groupby("сдал")[["часы", "посещаемость", "проектов"]].mean().round(1))
print(clean[["часы", "посещаемость", "проектов", "балл"]].corr().round(2))

X = clean[["часы", "посещаемость", "проектов"]]
y = clean["сдал"]
print(X.shape, y.shape)
```
