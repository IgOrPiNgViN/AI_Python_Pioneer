## Данные: подготовка к экзамену

Ученики 9 класса готовятся к экзамену по математике. Для каждого известно:

| Столбец | Смысл |
|---------|-------|
| `id` | номер ученика |
| `часы_подготовки` | часов подготовки в неделю |
| `пробников` | сколько пробных вариантов решено (из 10) |
| `сон` | часов сна в сутки |
| `репетитор` | 1 — занимается с репетитором, 0 — нет |
| `телефон` | часов в телефоне в сутки |
| `балл` | итоговый балл за экзамен (0–100) |
| `сдал_на_4` | `True`, если балл 60 и выше |

Запусти ячейку — она создаст таблицу `df`.

````python
import random
import pandas as pd

random.seed(2027)
rows = []
for i in range(240):
    hours = random.randint(0, 12)
    mocks = random.randint(0, 10)
    sleep = round(random.uniform(5.5, 9.5), 1)
    if random.random() < 0.35:
        tutor = 1
    else:
        tutor = 0
    phone = random.randint(1, 8)
    score = round(30 + 3 * hours + 2.5 * mocks + 2 * (sleep - 7) + 6 * tutor - 1.5 * phone + random.gauss(0, 8))
    rows.append({
        "id": i + 1,
        "часы_подготовки": hours,
        "пробников": mocks,
        "сон": sleep,
        "репетитор": tutor,
        "телефон": phone,
        "балл": min(100, max(0, score)),
    })
df = pd.DataFrame(rows)
df["сдал_на_4"] = df["балл"] >= 60
df.head()
````

| Вариант | Baseline | Модель |
|---------|----------|--------|

```python
feats = ["часы_подготовки", "пробников", "сон", "репетитор", "телефон"]
X = df[feats]

# Вариант К
y = df["сдал_на_4"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
dummy = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
print("baseline:", round(accuracy_score(y_test, dummy.predict(X_test)), 2))
tree = DecisionTreeClassifier(max_depth=3, random_state=0).fit(X_train, y_train)
pred = tree.predict(X_test)
print(confusion_matrix(y_test, pred))
print("acc", round(accuracy_score(y_test, pred), 2), "prec", round(precision_score(y_test, pred), 2),
      "rec", round(recall_score(y_test, pred), 2))
print("train", round(accuracy_score(y_train, tree.predict(X_train)), 2))

# Вариант Р
y = df["балл"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print("baseline MAE:", round(mean_absolute_error(y_test, DummyRegressor().fit(X_train, y_train).predict(X_test)), 1))
lin = LinearRegression().fit(X_train, y_train)
print("MAE test:", round(mean_absolute_error(y_test, lin.predict(X_test)), 1),
      "train:", round(mean_absolute_error(y_train, lin.predict(X_train)), 1))
print(pd.Series(lin.coef_, index=feats).round(2))
```

Коэффициенты линейной регрессии близки к заложенным: каждый час подготовки ≈ +3 балла, пробник ≈ +2.8, репетитор ≈ +7, час в телефоне ≈ −1.5.

**Хорошие ограничения:** обучена на 240 выдуманных учениках одной школы; признаки — самоотчёт (часы, сон, телефон можно указать неточно); нельзя использовать для выставления оценок или «отсева»; ошибка FN (не заметили риск) хуже — решение о помощи принимает учитель; «репетитор» связан с достатком семьи — модель может закреплять неравенство.
