## Часть B. Практика (16 баллов)

**Ситуация.** Учитель хочет **заранее**, за месяц до зачёта, понять, кто из учеников может его не сдать, чтобы позаниматься с ними дополнительно. Данные: часы занятий в неделю, сколько задач решено из сборника, сон, занимается ли с репетитором, итоговый балл за зачёт и сдан ли зачёт.

Запусти ячейку с данными, затем выполни пункты B1–B8.

````python
import random
import pandas as pd

random.seed(34)
rows = []
for i in range(200):
    study = random.randint(0, 12)
    tasks = random.randint(0, 20)
    sleep = round(random.uniform(5, 9.5), 1)
    if random.random() < 0.3:
        tutor = 1
    else:
        tutor = 0
    score = round(15 + 3.5 * study + 1.5 * tasks + 3 * (sleep - 7) + 8 * tutor + random.gauss(0, 9))
    rows.append({
        "id": i + 1,
        "часы_занятий": study,
        "решено_задач": tasks,
        "сон": sleep,
        "репетитор": tutor,
        "балл": min(100, max(0, score)),
    })
df = pd.DataFrame(rows)
df["зачёт"] = df["балл"] >= 50
df.head()
````
