## 1. Зачем этот курс

Вокруг всё больше систем, которые **распознают** лица, **советуют** видео, **переводят** текст, **пишут** ответы в чате. Но «поставить нейросеть» — это малая часть работы. Нужно ещё:

- понять, **какую задачу** решаем и как поймём, что решили хорошо;
- собрать и проверить **данные** (мусор на входе → мусор на выходе);
- **честно проверить**, что решение работает на новых примерах;
- сказать, **где оно ошибается** и кому может навредить.

**Инженерия ИИ** — это умение пройти весь этот путь осознанно.

> **Аналогия с мостом.** Можно купить отличные балки — это «готовая модель». Но без чертежа, расчёта нагрузки и проверки мост опасен. Инженер отвечает за весь мост, а не только за балки.

## 2. Что такое искусственный интеллект

**Искусственный интеллект (ИИ)** — это компьютерные системы, которые решают задачи, для которых обычно нужны «умные» способности человека.

| Способность | Пример задачи |
|-------------|---------------|
| Распознать | лицо на фото, речь, рукописную цифру |
| Понять текст | перевести, найти ответ, определить настроение отзыва |
| Предсказать | пробки, погоду, оценку за контрольную |
| Сгенерировать | текст, картинку, музыку, код |
| Посоветовать | фильм, песню, товар, маршрут |

Важные уточнения:

1. ИИ — **не** «мыслящий робот из фильма» с желаниями и сознанием. Это программы.
2. Система, сильная в одной задаче, может быть бесполезной в другой. Переводчик не умеет водить машину.
3. Уверенный и красивый ответ **не равен** правильному.

Ты наверняка слышал слова **нейросеть** и **языковая модель**. Нейросеть — один из видов моделей машинного обучения (что это такое — в следующем разделе); она особенно хороша для картинок, звука и текста. Языковая модель — нейросеть, которая умеет продолжать и писать текст; на ней работают чат-боты вроде ChatGPT или GigaChat. Подробно о них — в конце курса.

## 3. Обычная программа и машинное обучение — главное отличие урока

### Обычная программа (правила)

Человек **сам** придумывает и записывает логику:

> «Если температура выше 30 — написать „жарко“».

Программа ничему не учится. Она делает ровно то, что записано. Появился новый случай — программист дописывает новое правило.

### Машинное обучение (ML, *machine learning*)

Мы даём компьютеру много **примеров с ответами**, и он сам **находит закономерность**:

> Показали тысячи фото кошек и собак с подписями → модель научилась отличать их на новых фото.

Результат обучения называют **моделью**. Модель умеет давать ответ для примеров, которых раньше не видела.

| | Правила | Машинное обучение |
|--|---------|-------------------|
| Кто придумывает логику | человек | модель находит её в данных |
| Нужны ли примеры с ответами | обычно нет | да, много и хороших |
| Легко ли объяснить решение | обычно да | иногда трудно |
| Когда подходит | случаев мало, логика ясна | случаев много, логику трудно записать |
| Пример | скидка 10% при покупке больше 1000 ₽ | рекомендации видео |
| Главный риск | забыли редкий случай | плохие или однобокие данные |

> **Аналогия.** Правила — это точный рецепт: «2 ложки сахара, варить 10 минут». ML — это повар, который попробовал тысячу блюд с оценками «вкусно/невкусно» и сам понял, что работает.

В настоящих продуктах подходы часто **комбинируют**: модель предлагает решение, а правила страхуют от опасных ошибок.

### Частая ошибка

Путать «выглядит умно» и «обучено на данных». Программа, которая ставит отзыву оценку по словам «отлично» и «ужас», — это **правила**, даже если результат похож на умный. ML начинается там, где закономерность найдена по примерам, а не перечислена человеком вручную.

## 5. Где ИИ вокруг нас

Пройдись мысленно по своему дню. Где могла участвовать система, похожая на ИИ?

1. **Рекомендации** — видео, музыка, лента новостей, магазины.
2. **Голос** — голосовые помощники, автоматические субтитры, диктовка.
3. **Почта** — фильтр спама.
4. **Камера телефона** — разблокировка по лицу, улучшение фото.
5. **Переводчики** текста и речи.
6. **Чат-боты** и генерация текста и картинок.
7. **Навигатор** — прогноз пробок и времени в пути.
8. **Клавиатура** — автоисправление и подсказки слов.

**Честный инженерный ответ.** Снаружи мы не всегда знаем, что внутри: правила или обученная модель. Тогда правильно сказать «похоже на ML, но точно не знаю». Честная неуверенность лучше уверенной выдумки.

## 6. Что ИИ делает хорошо, а что — опасно без контроля

**Хорошо:**
- однотипная работа на огромных объёмах данных;
- поиск закономерностей, которые человек не успевает заметить;
- черновики и подсказки (текст, идеи, код).

**Опасно без контроля:**
- уверенно выдаёт неправду (у языковых моделей это называют **галлюцинациями**);
- повторяет ошибки и несправедливость, которые были в данных;
- решает **не ту** задачу, если её плохо поставили;
- создаёт ложное чувство: «раз умный — значит прав».

### Заблуждения

| Заблуждение | Как на самом деле |
|-------------|-------------------|
| «ИИ всегда прав» | Ошибается; его нужно проверять |
| «ИИ — это ChatGPT» | Чат-боты — лишь один вид ИИ-систем |
| «Красиво отвечает — значит понимает» | Может правдоподобно угадывать текст |
| «Важна модель, данные — мелочь» | Данные часто важнее модели |
| «С ИИ можно не думать самому» | Думать и проверять становится ещё важнее |

## 7. Кто такой инженер ИИ

Инженер ИИ проводит решение через **жизненный цикл**:

1. **Задача и критерий успеха.** Не «сделать ИИ», а «находить 9 из 10 спам-писем и почти не трогать нормальные».
2. **Данные.** Собрать и проверить примеры.
3. **Подход.** Правила, обучаемая модель или готовая модель.
4. **Обучение или настройка.**
5. **Честная проверка** на примерах, которых система раньше не видела.
6. **Демонстрация и ограничения.** Где ошибается? Кому может навредить? Когда нужен человек?

Этот цикл — карта всего нашего курса:

| Модуль | Уроки | Чему учимся |
|--------|------:|-------------|
| Старт и инструменты | 1–2 | что такое ИИ, как работать с ИИ-помощником |
| Python | 3–9 | язык, на котором работают с данными |
| Данные | 10–16 | таблицы, очистка, разведка, графики |
| Машинное обучение | 17–25 | модели, честная проверка, метрики, этика |
| ИИ-приложения | 26–29 | картинки, текст, большие языковые модели |
| Итоговый проект | 30–36 | весь цикл на своей задаче, защита |
