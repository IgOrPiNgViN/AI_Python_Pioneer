## 1. Что такое промпт

**ИИ-помощник** — чат-бот на основе языковой модели (урок 1): например, GigaChat, YandexGPT (Алиса), ChatGPT. Ты пишешь ему сообщение — он отвечает текстом.

**Промпт** — текст (иногда вместе с файлами или картинками), который ты отправляешь языковой модели.

> **Аналогия с кафе.** «Еды» — непонятно, что принести. «Чего-нибудь вкусного» — всё ещё гадание. «Какао без сахара, средний стакан, с собой» — понятно. Чем точнее заказ, тем больше шанс получить нужное.

**Промпт-инженерия** — умение:

1. ясно формулировать задачу;
2. давать нужный контекст;
3. просить нужный формат ответа;
4. ставить границы — чего делать не надо;
5. **улучшать** запрос, если ответ получился слабым.

Это не «волшебная фраза из интернета», а навык ясно мыслить и ясно просить.

## 2. Чат и агент

**Чат с языковой моделью.** Ты пишешь — модель отвечает текстом. Один вопрос — один ответ.

**Агент.** Может больше:

- разбить цель на **шаги**;
- пользоваться **инструментами**: поиском, запуском кода, чтением файлов;
- выполнить несколько шагов подряд и вернуться с результатом.

Поэтому агенту особенно важно знать:

- **цель** — что считать готовым результатом;
- **границы** — чего делать нельзя;
- **уровень** — ты ученик 8–9 класса и хочешь научиться, а не получить готовую работу.

Если границы не задать, агент может сделать всю работу сам — и ты ничему не научишься.

## 3. Схема сильного запроса

| Часть | Вопрос | Пример |
|-------|--------|--------|
| **Роль** | Кем должен быть помощник? | «Ты помощник ученика 8 класса» |
| **Контекст** | Что ему нужно знать? | условие задачи, текст ошибки, что ты уже пробовал |
| **Задача** | Что нужно **сейчас**? | одна конкретная задача лучше пяти |
| **Формат** | Каким должен быть ответ? | «3 пункта», «таблица», «коротко» |
| **Ограничения** | Чего делать не надо? | «не решай за меня», «простыми словами» |

### Шаблон курса (сохрани себе)

```text
Роль: ты помощник ученика 8–9 класса на курсе «Инженерия ИИ».
Контекст: <условие / текст ошибки / что я уже пробовал>
Задача: <что нужно прямо сейчас>
Формат: кратко, по пунктам.
Ограничения: простыми словами; не решай всё за меня —
дай подсказку и задай наводящий вопрос.
```

Почему порядок важен: модель продолжает твой текст. Если сначала сказать, что она помогает *учиться*, и запретить готовые решения, ответ будет полезнее для учёбы.

## 4. Демо: плохой и хороший запрос

Задача: понять, что такое машинное обучение (тема урока 1).

**Плохой запрос:**

```text
Расскажи про машинное обучение.
```

Что не так: непонятно, для кого, сколько и зачем. Скорее всего придёт длинная лекция со сложными словами.

**Хороший запрос:**

```text
Роль: ты помощник ученика 8 класса.
Контекст: на уроке мы обсуждали разницу между программой с правилами
и машинным обучением. Я понял про правила, но не до конца про ML.
Задача: объясни, чем ML отличается от правил, на примере из футбола.
Формат: 3–4 предложения и один пример.
Ограничения: без формул и сложных терминов; в конце задай мне
один вопрос, чтобы я проверил, понял ли.
```

Преподаватель покажет ответы на оба запроса. Сравните: какой ответ полезнее и почему?

## 5. Почему плохие запросы дают плохие ответы

| Проблема | Что происходит | Как исправить |
|----------|----------------|---------------|
| Слишком широко | модель гадает, о чём ты | сузить до одной задачи |
| Нет контекста | общий ответ «мимо» | добавить условие, текст ошибки, что уже пробовал |
| Нет формата | длинная «простыня» | попросить 3 пункта, таблицу, шаги |
| Нет ограничений | делает всё за тебя | «не решай за меня», «подсказка» |
| Всё сразу | путаница | разбить на 2–3 сообщения |
| Нет проверки | красивая ошибка | проверить сам: запустить, сверить с учебником |

## 6. Итерации: хороший ответ — часто со второй-третьей попытки

Новичок думает: «Спросил — не помогло — ИИ плохой». Инженер думает: «Уточню запрос».

Пример цепочки:

1. «Что значит SyntaxError?» — общий ответ.
2. «Я убрал кавычку в `print("Привет")` и получил SyntaxError. Почему?» — уже про твой случай.
3. «Объясни в двух предложениях и не исправляй код за меня — я хочу найти ошибку сам» — ответ, который помогает учиться.

Каждый шаг **сужает** задачу и делает ответ полезнее.

## 7. Как читать ответ помощника

Не копируй ответ сразу. Пройди чек-лист:

1. **Понял ли я каждое предложение?** Если нет — попроси объяснить проще.
2. **Можно ли это проверить?** Запуском кода, учебником, вторым источником.
3. **Нет ли лишнего?** Того, о чём ты не просил.
4. **Соблюдены ли ограничения?** Если просил подсказку, а пришло готовое решение — не используй его.
5. **Могу ли я пересказать ответ своими словами за 30 секунд?** Если нет — ответ ещё не стал твоим знанием.

## 8. Что можно и чего нельзя

| Можно | Нельзя |
|-------|--------|
| «Объясни ошибку в этой строке» | «Сделай всё задание за меня» |
| «Подскажи следующий шаг» | копировать ответ, не читая |
| «Объясни проще, на примере» | сдавать то, что не понимаешь |
| «Задай мне наводящий вопрос» | просить решить самостоятельную или контрольную |
| «Придумай похожую задачу для тренировки» | отправлять пароли, адреса, чужие фото и личные данные |

**Почему нельзя отправлять личные данные.** Всё, что ты отправляешь в чат, уходит на чужой сервер. Пароли, номера телефонов, фамилии одноклассников, фото людей без их согласия — не для чата. Инженеры в компаниях соблюдают то же правило.

> **Аналогия с навигатором.** Навигатор помогает доехать, но куда ехать, решаешь ты. Если отдать ему руль и смотреть в телефон — авария. Ты — пилот, ИИ — помощник со схемой маршрута.

### Практика 2. Запрос про ошибку (10 минут)

Запусти ячейку ниже. В ней специально пропущена закрывающая кавычка — ты увидишь сообщение об ошибке.

Строки, которые начинаются со знака `#`, — **комментарии**: это пояснения для человека, Python их пропускает. Выполнится только последняя строка с `print`.

````python
# Здесь намеренная ошибка: пропущена закрывающая кавычка.
# Запусти ячейку и посмотри на сообщение об ошибке.
# Потом допиши кавычку перед скобкой и запусти снова.
print("Ошибка — это подсказка, а не катастрофа)
````

Составь запрос к помощнику, который:

- объясняет **причину** ошибки простыми словами;
- **не** даёт исправленный код — ты хочешь исправить сам;
- в конце задаёт тебе **наводящий вопрос**.

Ошибка будет называться **SyntaxError** — «синтаксическая ошибка»: код записан не по правилам языка, как предложение без точки или с незакрытой скобкой. Python даже не начинает его выполнять.

После того как разберёшься, исправь ячейку выше: добавь кавычку и запусти снова.

**Мой запрос про ошибку:**

- Роль: …
- Контекст: (вставь строку кода и текст ошибки) …
- Задача: …
- Формат: …
- Ограничения: …

## Итог урока

С этого урока ИИ-помощником **можно** пользоваться на курсе — по этим правилам. На самостоятельных и контрольной режим объявляет преподаватель.
