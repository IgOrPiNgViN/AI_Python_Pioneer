## 1. Зачем Python инженеру ИИ

Инженер ИИ постоянно делает одно и то же:

1. **получает** значения (из файла, из таблицы, от пользователя);
2. **приводит** их к нужному виду;
3. **считает** (среднее, долю ошибок, прогноз);
4. **выводит** понятный результат.

Python — самый популярный язык для этого. Его создал Гвидо ван Россум в 1991 году с идеей: код должен легко **читаться человеком**. Сегодня мы освоим первые «кирпичики» этого языка.

## 2. Переменные

**Переменная** — это имя, к которому привязано значение. Как подписанная коробка: на коробке имя, внутри — значение.

```python
age = 14
```

- `age` — имя переменной;
- `=` — **присвоить** («положи в коробку»), а не «равно» из математики;
- `14` — значение.

Значение можно **заменить**: `age = 15`. Справа можно использовать старое значение: `age = age + 1` — сначала Python вычисляет правую часть, потом кладёт результат в коробку.

````python
name = "Аня"
age = 14
print(name)
print(age)

age = age + 1
print("Через год:", age)
````

Заметь разницу: `print(name)` — **без кавычек** — печатает то, что лежит в переменной (`Аня`). А `print("name")` — **в кавычках** — напечатал бы само слово `name`. Кавычки говорят Python: «это текст, а не имя переменной».

Через запятую в `print` можно передать несколько значений: `print("Через год:", age)`.

### Правила хороших имён

| Правило | Хорошо | Плохо |
|---------|--------|-------|
| Имя объясняет смысл | `score`, `user_name` | `a`, `x1`, `tmp2` |
| Без пробелов, слова через `_` | `total_minutes` | `total minutes` |
| Не начинается с цифры | `score2` | `2score` |
| Не совпадает с командами Python | `print_count` | `print`, `if`, `for` |
| Регистр важен | `Score` и `score` — **разные** переменные | |

Ясное имя лучше короткого: через неделю ты сам не вспомнишь, что хранится в `x`.

## 3. Четыре базовых типа

| Тип | Примеры | Что хранит |
|-----|---------|------------|
| `int` | `14`, `-3`, `0` | целое число |
| `float` | `1.65`, `3.0`, `-0.5` | дробное число (через **точку**!) |
| `str` | `"привет"`, `'14'` | текст, всегда в кавычках |
| `bool` | `True`, `False` | логическое «да / нет» |

### Что такое функция

`print` — это **функция**: готовая команда Python. Её вызывают так: имя, а в скобках — то, с чем ей работать (это называют **аргументом**). Многие функции не только что-то делают, но и **возвращают результат** — отдают значение, которое можно напечатать или положить в переменную. Например, `len("привет")` возвращает число 6. Своих функций мы научимся писать на уроке 7, а пока будем пользоваться готовыми.

Функция `type()` возвращает тип значения. Не гадай — проверяй.

````python
height = 1.65
is_student = True
lesson = "урок 3"

print(type(age))
print(type(height))
print(type(lesson))
print(type(is_student))
print(type("14"))   # в кавычках — значит строка, даже если внутри цифры
````

Python печатает тип в виде `<class 'int'>`: слово `class` здесь означает «вид значения», важна часть в кавычках — `int`, `float`, `str` или `bool`.

Обрати внимание: `14` и `"14"` — **разные** вещи. Первое — число, с ним можно считать. Второе — текст из двух символов.

`True` и `False` пишутся с большой буквы. Подробно тип `bool` понадобится на следующем уроке, когда программа начнёт принимать решения.

## 4. Арифметика

| Операция | Знак | Пример | Результат |
|----------|:----:|--------|-----------|
| сложение | `+` | `7 + 2` | `9` |
| вычитание | `-` | `7 - 2` | `5` |
| умножение | `*` | `7 * 2` | `14` |
| деление | `/` | `7 / 2` | `3.5` — всегда `float` |
| целая часть от деления | `//` | `7 // 2` | `3` |
| остаток от деления | `%` | `7 % 2` | `1` |
| степень | `**` | `2 ** 3` | `8` |

Порядок действий как в математике: сначала `**`, потом `* / // %`, потом `+ -`. Скобки меняют порядок.

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
print(2 + 3 * 4)
print((2 + 3) * 4)
````

### Зачем `//` и `%`

Они нужны, когда что-то делят «на целые части»: минуты на часы, ученики на группы, страницы на тетради.

**Пример:** 135 минут — это сколько часов и минут?

````python
total_minutes = 135
hours = total_minutes // 60     # сколько целых часов
minutes = total_minutes % 60    # сколько минут осталось
print(hours, "ч", minutes, "мин")
````

**Попробуй сам.** Поменяй `total_minutes` на `59`, `60`, `250` и запусти снова. Совпадает ли результат с тем, что ты посчитал в уме?

### Округление

Результат деления бывает длинным. Функция `round(число, знаков)` округляет:

````python
price = 1999
discount = price * 15 / 100
print(discount)
print(round(discount, 2))
print(round(3.7))   # без второго числа — до целого
````

## 5. Строки

Строка — это текст в кавычках. С ней тоже можно делать операции:

| Действие | Пример | Результат |
|----------|--------|-----------|
| склеить | `"Инженерия" + " ИИ"` | `"Инженерия ИИ"` |
| повторить | `"ха" * 3` | `"хахаха"` |
| длина | `len("привет")` | `6` — число символов |

**Нельзя** склеить строку и число: `"Урок " + 3` даст ошибку `TypeError`. Сначала превращаем число в строку: `str(3)`.

````python
first = "Инженерия"
second = " ИИ"
print(first + second)
print("ха" * 3)
print(len(first))

lesson_number = 3
print("Урок " + str(lesson_number))
````

### f-строки — удобный способ вставить значения в текст

Перед кавычкой ставим букву `f`, а переменные пишем в фигурных скобках:

````python
name = "Ира"
age = 14
print(f"Привет, {name}! Тебе {age} лет.")
print(f"Через 5 лет тебе будет {age + 5}.")
````

## 6. Вывод: `print`

- Несколько значений через запятую печатаются через пробел.
- `sep=` — чем разделить значения.
- `end=` — чем закончить строку (обычно это перенос строки).

````python
print("Курс:", "Инженерия ИИ", "урок", 3)
print("2026", "10", "08", sep="-")
print("Загрузка...", end=" ")
print("готово!")
````

## 7. Ввод: `input` — главная ловушка урока

`input("вопрос")` показывает вопрос и ждёт, пока пользователь что-то напечатает и нажмёт Enter. В Colab под ячейкой появится поле для ввода. Пока ты не введёшь ответ, ячейка «крутится» — это нормально. То, что ввёл пользователь, `input` возвращает — поэтому результат кладут в переменную: `a = input(...)`.

**Самое важное:** `input` **всегда** возвращает строку — даже если ввели цифры.

Запусти ячейку и введи два числа, например `3` и `5`.

````python
a = input("Первое число: ")
b = input("Второе число: ")
print("Тип a:", type(a))
print("a + b =", a + b)
````

Получилось `35`, а не `8`! Строки `"3"` и `"5"` **склеились**. Чтобы считать, нужно **преобразовать тип**:

| Функция | Что делает | Пример | Осторожно |
|---------|------------|--------|-----------|
| `int(x)` | в целое число | `int("42")` → `42` | `int("3.5")` — ошибка |
| `float(x)` | в дробное число | `float("3.5")` → `3.5` | `float("три")` — ошибка |
| `str(x)` | в строку | `str(100)` → `"100"` | почти всегда работает |

Функции можно вкладывать друг в друга: `int(input("Первое число: "))`. Python выполняет их **изнутри наружу**: сначала `input` получает текст `"3"`, потом `int` превращает его в число `3`.

````python
a = int(input("Первое число: "))
b = int(input("Второе число: "))
print("a + b =", a + b)
````

Ещё одна ловушка: `int(3.9)` **отбрасывает** дробную часть и даёт `3`, а не округляет. Для округления — `round(3.9)`.

````python
print(int(3.9))
print(round(3.9))
print(float("3.5") + 1)
print(int(float("3.5")))
````

### Почему это важно для работы с данными

Данные из файлов и таблиц тоже часто приходят **текстом**. Если не превратить `"4"`, `"5"`, `"3"` в числа, вместо среднего получится склейка `"453"`. Программа не упадёт — она тихо посчитает ерунду. Это опаснее явной ошибки.

## 8. Как читать ошибки

Когда что-то не так, Python показывает **сообщение об ошибке**. Читай его **снизу вверх**: в последней строке — тип ошибки и пояснение.

| Ошибка | Типичная причина сегодня |
|--------|--------------------------|
| `NameError` | опечатка в имени: `Print` вместо `print`, переменная не создана |
| `TypeError` | сложили строку и число: `"Урок " + 3` |
| `ValueError` | `int("3.5")`, `int("привет")` |
| `ZeroDivisionError` | деление на ноль |
| `SyntaxError` | забыли кавычку или скобку |

Ниже — строки с ошибками, они закомментированы (символ `#` в начале строки — это **комментарий**, Python его не выполняет). Убирай `#` **по одной строке**, запускай, читай ошибку, потом возвращай `#`.

````python
# print("Урок " + 3)
# print(int("3.5"))
# Print("привет")
# print(10 / 0)
print("Все ошибки закомментированы — ячейка работает.")
````
