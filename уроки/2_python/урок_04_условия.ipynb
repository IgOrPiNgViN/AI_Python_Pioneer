## 1. Сравнения и тип `bool`

Сравнение — это вопрос, на который Python отвечает `True` (да) или `False` (нет).

| Оператор | Смысл | Пример | Результат |
|:--------:|-------|--------|-----------|
| `==` | равно | `5 == 5` | `True` |
| `!=` | не равно | `5 != 3` | `True` |
| `>` / `<` | больше / меньше | `5 > 3` | `True` |
| `>=` / `<=` | больше или равно / меньше или равно | `5 <= 3` | `False` |

**Не путай:** `=` — **присвоить** (положить в переменную), `==` — **сравнить**.

````python
score = 75
print(score > 60)
print(score == 100)
print(score != 100)

passed = score >= 60      # результат сравнения можно сохранить в переменную
print("Сдал?", passed, type(passed))
````

Строки сравниваются **с учётом регистра**: `"ИИ" == "ии"` — это `False`.

Ещё одна ловушка из урока 3: строки с цифрами сравниваются **по символам**, а не как числа.

````python
print("ИИ" == "ии")
print("10" > "9")    # сравнение строк: символ "1" меньше символа "9"
print(10 > 9)        # сравнение чисел
````

Поэтому числа из `input` **всегда** сначала превращаем в `int` или `float` — и только потом сравниваем.

## 2. `if` — «если»

```python
if условие:
    действие
```

- после условия — **двоеточие** `:`;
- строки внутри ветки пишутся с **отступом** (4 пробела; Colab ставит его сам после `:`);
- код **без** отступа после блока выполняется всегда.

Отступ в Python — не украшение, а часть языка. Он показывает, какие строки относятся к ветке.

````python
score = 95

if score >= 90:
    print("Отлично!")
    print("Эта строка тоже внутри ветки")

print("Проверка закончена")   # без отступа — выполняется всегда
````

**Попробуй сам.** Поменяй `score` на `50` и запусти. Какие строки пропали?

## 3. `if / else` — «если… иначе…»

`else` срабатывает, когда условие ложно. У `else` нет своего условия.

````python
temperature = 12

if temperature >= 18:
    print("Можно идти без куртки")
else:
    print("Надень куртку")
````

## 4. `if / elif / else` — несколько вариантов

`elif` означает «иначе, если». Python проверяет условия **сверху вниз** и выполняет **первую** ветку, условие которой истинно. Остальные пропускаются.

````python
score = 75   # попробуй: 95, 75, 60, 59, 30

if score >= 90:
    print("Отлично")
elif score >= 60:
    print("Зачёт")
else:
    print("Нужно повторить")
````

### Порядок веток важен

Посмотри на код ниже. Пороги те же, но порядок другой. Что выведется для `score = 95`?

````python
score = 95

if score >= 60:
    print("Зачёт")
elif score >= 90:
    print("Отлично")       # сюда при таком порядке не попасть никогда
else:
    print("Нужно повторить")
````

Выводится «Зачёт»: 95 ≥ 60 — первая ветка уже истинна, и до проверки `>= 90` дело не доходит.

**Правило:** более строгие условия ставим **выше**. Такие ошибки называют **логическими**: программа не падает, но решает неправильно. Они опаснее явных ошибок, потому что их не видно.

## 5. Тест-кейсы: как проверить все ветки

Инженер не верит, что код работает, пока не проверил. Для условий составляют **таблицу тестов**:

- хотя бы по одному значению на **каждую ветку**;
- значения **на границах** порогов (ровно 60, чуть меньше — 59).

| Вход `score` | Ожидаем | Получили | Совпало? |
|-------------:|---------|----------|:--------:|
| 95 | Отлично | | |
| 90 | Отлично | | |
| 89 | Зачёт | | |
| 60 | Зачёт | | |
| 59 | Нужно повторить | | |

Пока проверяем вручную: меняем значение и перезапускаем ячейку. На следующем уроке научимся прогонять все тесты одним циклом.

## 6. `and`, `or`, `not`

| Операция | Значит | Истинно, когда… |
|----------|--------|-----------------|
| `A and B` | и | **оба** условия истинны |
| `A or B` | или | **хотя бы одно** истинно |
| `not A` | не | `A` ложно |

````python
age = 15
has_ticket = True

if age >= 14 and has_ticket:
    print("Можно войти")
else:
    print("Вход закрыт")

priority = "high"
errors = 1
if priority == "high" or errors > 3:
    print("Заявку смотрит человек")
else:
    print("Заявку можно обработать автоматически")

is_weekend = False
if not is_weekend:
    print("Сегодня учимся")
````

**Частая ошибка:** думать, что `or` означает «оба». Нет: для `or` достаточно одного.

Заметь: в `if age >= 14 and has_ticket:` вторая часть — не сравнение, а просто переменная. Это можно, потому что в `has_ticket` уже лежит `True` или `False`. Писать `has_ticket == True` не нужно. Так же `if not is_weekend:` читается как «если не выходной».

Полезный приём: число **чётное**, если остаток от деления на 2 равен нулю: `n % 2 == 0`.

### Длинные условия разбивай на понятные части

Сложное условие в одну строку трудно читать и проверять. Сохрани части в переменные с понятными именами:

````python
errors = 4
priority = "low"

too_many_errors = errors > 3
is_urgent = priority == "high"
needs_human = too_many_errors or is_urgent

print("Много ошибок:", too_many_errors)
print("Срочно:", is_urgent)
print("Нужен человек:", needs_human)
````

## 7. Оператор `in` для строк

`часть in строка` проверяет, есть ли кусок текста внутри строки. Это удобно для простых фильтров.

````python
message = "Купи сейчас! Скидка только сегодня"

if "скидка" in message:
    print("Подозрение на рекламу")
else:
    print("Похоже на обычное сообщение")
````

Ничего не нашлось? Всё правильно: в сообщении «Скидка» с большой буквы, а ищем «скидка». Регистр важен! Как приводить текст к одному регистру, узнаем на уроке 6.

## 9. Правила и машинное обучение

Условия — это **правила**, которые написал человек. Помнишь урок 1? Это ещё **не** машинное обучение.

| Правила (`if`) | Машинное обучение |
|----------------|-------------------|
| Порог выбрал человек | Закономерность найдена по примерам |
| Легко объяснить «почему» | Иногда трудно объяснить |
| Новый случай — надо дописать правило | Может обобщать на новые случаи |

В реальных системах их часто **соединяют**: модель выдаёт уверенность от 0 до 1, а правило решает, что делать дальше.

````python
model_confidence = 0.62   # «уверенность модели», что комментарий грубый

if model_confidence >= 0.9:
    action = "скрыть комментарий"
elif model_confidence >= 0.6:
    action = "отправить модератору"
else:
    action = "опубликовать"

print("Решение:", action)
````

Модель здесь — только источник числа. Решение принимает **правило**, которое написали люди. Пороги `0.9` и `0.6` — тоже выбор инженера, и за него он отвечает.

## 10. Частые ошибки

| Ошибка | Что видно | Как чинить |
|--------|-----------|------------|
| Забыли `:` после условия | `SyntaxError` | добавить `:` |
| Нет отступа в ветке | `IndentationError` | 4 пробела |
| `if x = 10:` | `SyntaxError` | сравнение — это `==` |
| Сравнили строку из `input` с числом | `TypeError` | сначала `int()` / `float()` |
| Неправильный порядок `elif` | тихая логическая ошибка | строгие условия выше, тест-кейсы |
