## 1. Зачем нужны циклы

Нужно напечатать «Привет» 5 раз. Можно так:

```python
print("Привет")
print("Привет")
print("Привет")
print("Привет")
print("Привет")
```

А если 1000 раз? А если в каждой строке нужно что-то посчитать, и в расчёте нашлась ошибка — исправлять в тысяче мест? **Цикл** говорит компьютеру: «повтори этот блок кода».

| Цикл | Когда удобен | Мысль |
|------|--------------|-------|
| `for` | знаем, **по чему** идти: числа, символы, список | «для каждого…» |
| `while` | знаем **условие**, но не знаем число шагов | «пока…» |

## 2. Цикл `for` и `range`

```python
for i in range(5):
    print(i)
```

- `range(5)` — числа `0, 1, 2, 3, 4` (пять чисел, **начиная с нуля**, 5 не входит);
- переменная `i` на каждом шаге получает следующее число;
- тело цикла — строки с **отступом**, как в `if`.

| Запись | Какие числа | Комментарий |
|--------|-------------|-------------|
| `range(5)` | 0, 1, 2, 3, 4 | от 0 до 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | от 1, а 6 не входит |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 | с шагом 2 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 | обратный отсчёт |

````python
for i in range(5):
    print("Шаг", i)

print("Цикл закончился")   # без отступа — после цикла
````

````python
# Таблица умножения на 7. Чтобы 10 вошло, пишем range(1, 11)
n = 7
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
````

**Частая ошибка:** ждать, что `range(1, 10)` дойдёт до 10. Правая граница **не входит** — это нужно запомнить сразу.

## 3. Паттерн «копилка»

Чтобы посчитать сумму, заводим переменную-копилку **до** цикла, пополняем её **в** цикле и используем результат **после** цикла.

````python
total = 0                       # 1. копилка до цикла
for number in range(1, 6):      # числа 1, 2, 3, 4, 5
    total = total + number      # 2. пополняем
    print("добавили", number, "→ в копилке", total)
print("Сумма:", total)          # 3. результат после цикла
````

Запись `total = total + number` можно сократить: `total += number`. Так же работают `-=`, `*=`.

Копилкой можно и **считать**: `count += 1`, когда выполняется условие.

````python
count = 0
for number in range(1, 21):
    if number % 3 == 0:         # делится на 3 без остатка
        count += 1
print("Чисел от 1 до 20, которые делятся на 3:", count)
````

## 5. Списки — первое знакомство

Данные редко бывают одним числом. Оценки класса, температуры за неделю, длины сообщений — это **наборы** значений. В Python для этого есть **список**:

```python
scores = [4, 5, 3, 5, 4]
```

- элементы пишутся в **квадратных скобках** через запятую;
- в списке могут быть числа, строки и другие значения;
- `len(scores)` — сколько элементов;
- `[]` — пустой список.

Сегодня нам хватит трёх действий: **создать**, **перебрать в цикле**, **добавить** элемент через `append`. Подробно со списками будем работать на следующем уроке.

````python
scores = [4, 5, 3, 5, 4]
print(scores)
print("Сколько оценок:", len(scores))

for score in scores:
    print("Оценка:", score)
````

````python
fruits = []                     # пустой список
fruits.append("яблоко")         # добавить в конец
fruits.append("груша")
fruits.append("банан")
print(fruits, "— элементов:", len(fruits))
````

Обрати внимание на запись **через точку**: `fruits.append("яблоко")`. Так вызывают **метод** — функцию, которая «принадлежит» значению. `append` есть у каждого списка и добавляет элемент именно в **этот** список. Других методов будет много на уроке 6.

### Четыре главных паттерна обработки данных

Эти паттерны встречаются почти в каждой программе, которая работает с данными. Запомни их.

````python
scores = [4, 5, 3, 5, 2, 4, 5]

# 1. Сумма и среднее
total = 0
for s in scores:
    total += s
average = total / len(scores)
print("Среднее:", round(average, 2))

# 2. Сколько элементов подходят под условие
count_five = 0
for s in scores:
    if s == 5:
        count_five += 1
print("Пятёрок:", count_five)

# 3. Фильтр: собрать новый список из подходящих элементов
good = []
for s in scores:
    if s >= 4:
        good.append(s)
print("Оценки 4 и 5:", good)

# 4. Максимум: «лучший на данный момент»
best = 0
for s in scores:
    if s > best:
        best = s
print("Лучшая оценка:", best)
````

Почему в паттерне 4 копилка начинается с 0? Оценки всегда больше нуля, поэтому первая же оценка окажется больше и заменит 0. Для **минимума** начинают с заведомо **большого** числа (например, `worst = 1000`) и меняют знак сравнения на `<`.

В Python есть готовые функции `sum()`, `max()`, `min()`. Ими удобно **проверять** свои циклы:

````python
print(sum(scores), max(scores), min(scores))
````

Зачем тогда писать циклы самим? Потому что готовой функции нет для каждой задачи. «Сколько учеников сдали», «какие сообщения длиннее 100 символов», «среднее без пропусков» — всё это паттерны выше.

**Внимание:** если список пустой, `len(scores)` равно 0, и среднее посчитать нельзя — будет деление на ноль. Проверяй это условием.

## 6. Тест-кейсы одним циклом

На прошлом уроке мы проверяли условие, вручную меняя значение. Теперь можно прогнать **все** тесты сразу:

````python
test_scores = [95, 90, 89, 60, 59, 0]

for score in test_scores:
    if score >= 90:
        result = "Отлично"
    elif score >= 60:
        result = "Зачёт"
    else:
        result = "Нужно повторить"
    print(score, "→", result)
````

## 7. Цикл `while`

```python
while условие:
    действие
```

`while` повторяет блок, **пока** условие истинно. Удобен, когда заранее неизвестно, сколько будет шагов.

**Правило безопасности.** Перед запуском ответь себе: какая переменная стоит в условии и **где в цикле она меняется**? Если нигде — цикл никогда не закончится (**бесконечный цикл**).

````python
savings = 0
week = 0
while savings < 1000:           # копим на подарок 1000 ₽
    savings += 150              # каждую неделю откладываем 150 ₽
    week += 1
    print("Неделя", week, "— накоплено", savings)
print("Цель достигнута за", week, "недель")
````

## 8. `break` и `continue`

| Команда | Что делает | Пример из жизни |
|---------|------------|-----------------|
| `break` | сразу **выйти** из цикла | нашёл нужную книгу — ушёл из библиотеки |
| `continue` | пропустить остаток шага и перейти к **следующему** | бракованное яблоко — к следующему |

````python
# continue: пропускаем отсутствующие данные (-1 = «нет оценки»)
marks = [5, -1, 4, -1, 3]
for m in marks:
    if m == -1:
        continue
    print("Оценка:", m)

# break: ищем первое число больше 10
numbers = [3, 8, 12, 5, 20]
for x in numbers:
    if x > 10:
        print("Первое число больше 10:", x)
        break
````

## 9. Модуль `random` и игра «Угадай число»

Python умеет многое «из коробки», но часть возможностей хранится в **модулях**. Модуль подключают командой `import`. Модуль `random` умеет выбирать случайные числа:

Функции модуля вызывают **через точку**: `модуль.функция(...)`.

- `random.randint(1, 10)` — случайное целое от 1 до 10 (оба края входят);
- `random.choice(список)` — случайный элемент списка.

````python
import random

print(random.randint(1, 10))
print(random.choice(["камень", "ножницы", "бумага"]))
````

Игра: компьютер загадывает число от 1 до 10, у игрока 5 попыток. Цикл закончится либо когда угадали (`break`), либо когда попытки кончились — поэтому он не бесконечный.

````python
import random

secret = random.randint(1, 10)
attempts = 0

while attempts < 5:
    guess = int(input("Твоё число от 1 до 10: "))
    attempts += 1
    if guess == secret:
        print("Угадал за", attempts, "попыток!")
        break
    elif guess < secret:
        print("Загаданное число больше")
    else:
        print("Загаданное число меньше")

print("Загаданное число было:", secret)
````

## 10. Частые ошибки

| Ошибка | Что происходит | Как избежать |
|--------|----------------|--------------|
| В `while` не меняется переменная условия | бесконечный цикл | найди строку, которая её меняет |
| `range(1, 10)` вместо `range(1, 11)` | последнее число потерялось | правая граница не входит |
| Нет отступа в теле цикла | строка выполняется один раз после цикла | проверь отступы |
| Копилка создана **внутри** цикла | сумма каждый раз обнуляется | `total = 0` — до цикла |
| Среднее пустого списка | `ZeroDivisionError` | проверить `len(...) == 0` |

````python
raw = [21.5, -999, 22.0, 19.8, -999, 23.1, 18.0]
clean = []
dropped = 0
# TODO: цикл по raw; -999 не добавляем в clean, а считаем в dropped
````

<details>
<summary>Решение</summary>

```python
clean = []
dropped = 0
for value in raw:
    if value == -999:
        dropped += 1
        continue
    clean.append(value)

print("Чистые данные:", clean)
print("Выброшено:", dropped)
if len(clean) == 0:
    print("Нет данных для среднего")
else:
    print("Среднее:", round(sum(clean) / len(clean), 2))
```
</details>

Это первый маленький **конвейер обработки данных**: пройти → проверить → отфильтровать → посчитать. Ровно так работают с данными перед обучением моделей.
