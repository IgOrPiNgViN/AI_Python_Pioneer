## 1. Индексы: номер позиции

У каждого символа строки и каждого элемента списка есть **индекс** — номер позиции. Счёт идёт **с нуля**.

| Символ | П | р | и | в | е | т |
|--------|---|---|---|---|---|---|
| Индекс | 0 | 1 | 2 | 3 | 4 | 5 |
| С конца | -6 | -5 | -4 | -3 | -2 | -1 |

Отрицательные индексы считают с конца: `-1` — последний элемент.

````python
word = "Привет"
print(word[0])     # первый символ
print(word[5])     # последний
print(word[-1])    # тоже последний — удобно, если длина неизвестна

scores = [4, 5, 3, 5, 2]
print(scores[0], scores[-1])
````

Если индекса нет, будет ошибка `IndexError`. У строки из 6 символов индексы от 0 до 5, поэтому `word[6]` не существует.

````python
# Раскомментируй, прочитай ошибку и верни # на место:
# print(word[6])
print("Длина:", len(word), "→ последний индекс:", len(word) - 1)
````

## 2. Срезы: кусок последовательности

`a[start:stop]` — элементы с индекса `start` до `stop`, **не включая** `stop` (как в `range`).

| Срез | Что получится для `"Привет"` |
|------|------------------------------|
| `word[0:3]` | `"При"` |
| `word[2:]` | `"ивет"` — от 2 до конца |
| `word[:4]` | `"Прив"` — от начала до 3 |
| `word[::-1]` | `"тевирП"` — задом наперёд |

Третье число в срезе — **шаг**: `a[start:stop:step]`, как в `range`. `word[::2]` — каждый второй символ, а шаг `-1` идёт с конца, поэтому `word[::-1]` переворачивает строку. Пустое место в срезе значит «от самого начала» или «до самого конца».

````python
print(word[0:3])
print(word[2:])
print(word[:4])
print(word[::-1])

scores = [4, 5, 3, 5, 2, 4]
print("Первые три:", scores[:3])
print("Последние два:", scores[-2:])
````

## 3. Строки нельзя изменить, списки — можно

Это важное отличие:

- **список изменяемый**: `scores[0] = 100` заменит первый элемент;
- **строка неизменяемая**: `word[0] = "п"` даст ошибку. Нужно собрать **новую** строку.

````python
scores = [4, 5, 3]
scores[0] = 5
print(scores)

word = "кот"
# word[0] = "л"        # TypeError: строку нельзя менять по частям
new_word = "л" + word[1:]
print(word, "→", new_word)
````

## 4. Методы строк: чистим текст

**Метод** — это функция, которая «принадлежит» значению. Её вызывают через точку: `текст.lower()`.

| Метод | Что делает | Пример |
|-------|------------|--------|
| `strip()` | убирает пробелы по краям | `"  да ".strip()` → `"да"` |
| `lower()` / `upper()` | все буквы маленькие / большие | `"ДА".lower()` → `"да"` |
| `replace(a, b)` | заменяет `a` на `b` | `"да!".replace("!", "")` → `"да"` |
| `startswith(x)` / `endswith(x)` | начинается / заканчивается на `x`? | `"file.csv".endswith(".csv")` → `True` |
| `count(x)` | сколько раз встречается `x` | `"банан".count("а")` → `2` |
| `isdigit()` | состоит ли строка только из цифр? | `"42".isdigit()` → `True` |

**Главное правило:** методы строк **не меняют** исходную строку, а **возвращают новую**. Результат нужно сохранить: `text = text.strip()`.

````python
raw = "  Привет, МИР!  "
print(repr(raw))                  # repr показывает строку вместе с пробелами и кавычками
print(repr(raw.strip()))
print(repr(raw.strip().lower()))
print(repr(raw.strip().lower().replace("!", "")))

# Ловушка: забыли сохранить результат
s = "  hi  "
s.strip()
print(repr(s))       # пробелы на месте!
s = s.strip()
print(repr(s))       # теперь чисто
````

Функция `repr` показывает строку «как есть»: с кавычками по краям, поэтому видно лишние пробелы. Обычный `print` пробелы по краям не покажет.

Цепочка методов читается **слева направо**: сначала обрезали края, потом привели к нижнему регистру, потом убрали `!`. Каждый метод работает с результатом предыдущего.

## 5. `split` и `join`: строка ↔ список

- `текст.split()` — разрезать по пробелам и получить **список слов**;
- `текст.split(",")` — разрезать по запятой;
- `" ".join(список)` — склеить список строк в одну строку через пробел.

Запись `join` непривычная: метод вызывается у **строки-разделителя**, а список передаётся в скобках. `"-".join(words)` читается так: «соедини слова, ставя между ними `-`».

````python
sentence = "Инженерия ИИ учит проверять данные"
words = sentence.split()
print(words)
print("Слов:", len(words))
print("-".join(words))

line = "Ира,14,5"
parts = line.split(",")
print(parts)
````

### Строка чисел → список чисел

Данные часто приходят строкой: `"5 4 3 5"`. После `split` получаем список **строк** `['5', '4', '3', '5']`. Чтобы считать, каждую нужно превратить в число — циклом.

````python
line = "5 4 3 5"
numbers = []
for part in line.split():
    numbers.append(int(part))
print(numbers)
print("Среднее:", sum(numbers) / len(numbers))
````

## 6. Нормализация ответов пользователя

Люди пишут одно и то же по-разному: `Да`, `да`, ` ДА `, `Да!`. Без чистки условие `answer == "да"` пропустит большинство вариантов. Приводим текст к **единому виду** — это называется **нормализация**.

````python
answers = [" Да ", "ДА", "да!", "  Нет. ", "возможно", "Да"]

yes_count = 0
for answer in answers:
    clean = answer.strip().lower().replace("!", "").replace(".", "")
    print(repr(answer), "→", repr(clean))
    if clean == "да":
        yes_count += 1

print("Ответов «да»:", yes_count)
````

### Подстрока или целое слово?

Оператор `in` со строкой ищет **кусок текста** — даже внутри другого слова. Со списком `in` проверяет **целый элемент**. Для слов это важно:

````python
text = "мне это понятно"
print("нет" in text)            # True — нашлось внутри слова «понятно»!
words = text.split()
print("нет" in words)           # False — такого слова нет
````

Поэтому для поиска слов текст сначала разрезают на слова (их называют **токенами**), а потом проверяют каждое слово целиком.

## 7. Методы списков

| Действие | Код | Результат для `a = [3, 1, 2]` |
|----------|-----|-------------------------------|
| добавить в конец | `a.append(5)` | `[3, 1, 2, 5]` |
| вставить по индексу | `a.insert(0, 9)` | `[9, 3, 1, 2]` |
| удалить значение | `a.remove(1)` | `[3, 2]` |
| достать последний | `a.pop()` | вернёт `2` (его можно сохранить в переменную), станет `[3, 1]` |
| отсортировать сам список | `a.sort()` | `[1, 2, 3]` |
| получить отсортированную копию | `sorted(a)` | `[1, 2, 3]`, а `a` не меняется |
| есть ли элемент | `2 in a` | `True` |

````python
grades = [3, 5, 4, 2, 5]
grades.append(4)
print(grades)
print(sorted(grades))                   # копия по возрастанию
print(sorted(grades, reverse=True))     # копия по убыванию
print(grades)                           # сам список не изменился

grades.sort()
print(grades)                           # а теперь изменился

print(5 in grades, 1 in grades)
````

`reverse=True` — **именованный параметр**, как `sep=` у `print` из урока 3: дополнительная настройка функции. Здесь он означает «по убыванию».

## 8. `enumerate`: элемент вместе с номером

Иногда нужен не только элемент, но и его номер: «ошибка в третьей строке». Функция `enumerate` выдаёт пары **номер, элемент**. Поэтому в цикле сразу **две** переменные через запятую: в первую попадает номер, во вторую — сам элемент.

````python
tasks = ["собрать данные", "почистить", "обучить модель"]
for i, task in enumerate(tasks):
    print(i, task)

print("Нумерация с 1:")
for number, task in enumerate(tasks, start=1):
    print(f"{number}. {task}")
````

## 9. Частые ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|---------------|
| Забыли сохранить результат `strip()` / `lower()` | строка осталась грязной | `s = s.strip()` |
| Индекс за пределами | `IndexError` | индексы от 0 до `len - 1` |
| Ждали числа после `split` | получили список строк | `int()` для каждого элемента |
| Искали слово через `in` в строке | нашлось внутри другого слова | сначала `split()` |
| Пытаемся изменить символ строки | `TypeError` | собрать новую строку |
| Перепутали `sort()` и `sorted()` | список «не отсортировался» или «пропал» | `sort()` меняет список, `sorted()` даёт копию |

### Мини-проект. Тональность отзыва по словам (10 минут)

Определим, хороший отзыв или плохой, по словам-маркерам. Это **правила**, а не машинное обучение — но так работали первые системы анализа текста. Позже на курсе сравним правила с обученной моделью.

````python
positive = ["отлично", "супер", "нравится", "класс", "полезно", "понятно"]
negative = ["ужас", "плохо", "скучно", "непонятно", "запутанно"]

review = "Курс супер, но одно задание объяснено плохо!"

# 1. Нормализация: нижний регистр, убираем знаки препинания
clean = review.lower()
for mark in ",.!?":
    clean = clean.replace(mark, " ")
words = clean.split()
print("Слова:", words)

# 2. Считаем маркеры
pos = 0
neg = 0
for w in words:
    if w in positive:
        pos += 1
    if w in negative:
        neg += 1
print("Позитивных:", pos, "| негативных:", neg)

# 3. Решение по правилу
if pos > neg:
    print("Скорее позитивный отзыв")
elif neg > pos:
    print("Скорее негативный отзыв")
else:
    print("Непонятно / нейтрально")
````

- «Не понятно ничего» — после `split` получаются слова «не» и «понятно»; «понятно» есть в позитивных словах, и правило решает «позитивный». Правило не понимает **отрицания**: «не» меняет смысл соседнего слова.
- «Ну отлично, опять всё сломалось» — слово «отлично» делает отзыв «позитивным», хотя это сарказм. Слова «сломалось» нет в списке.

Ограничения: правила видят только слова из списков, не понимают отрицаний, сарказма и новых слов. Расширять списки можно бесконечно — люди всегда придумают новую формулировку.
