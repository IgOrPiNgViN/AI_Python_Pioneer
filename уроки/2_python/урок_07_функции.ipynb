## 1. Зачем нужны функции

Представь: среднее оценок нужно посчитать в пяти местах программы. Ты копируешь один и тот же цикл пять раз. Потом находишь ошибку — забыл проверку на пустой список. Исправлять придётся в пяти местах, и в одном ты обязательно забудешь. Отчёт будет врать.

С функцией логика записана **один раз**: исправил в одном месте — исправилось везде.

> **Аналогия.** Функция — это рецепт на карточке. Один раз записали рецепт соуса — и готовим его для пасты, пиццы и салата, подставляя разные продукты.

## 2. Как объявить и вызвать функцию

```python
def greet(name):
    message = f"Привет, {name}!"
    return message
```

| Часть | Смысл |
|-------|-------|
| `def` | «объявляю функцию» (*define*) |
| `greet` | имя функции |
| `(name)` | **параметр** — вход функции |
| `:` и отступ | тело функции, как у `if` и `for` |
| `return` | отдать результат и закончить функцию |

**Объявление** только учит Python рецепту — ничего не происходит. Чтобы функция сработала, её нужно **вызвать**: `greet("Ира")`.

````python
def greet(name):
    message = f"Привет, {name}!"
    return message

text = greet("Ира")          # вызов: результат сохранили в переменную
print(text)
print(greet("Максим"))       # результат сразу напечатали

for person in ["Аня", "Боря", "Катя"]:
    print(greet(person))
````

Значение, которое передают при вызове (`"Ира"`), называют **аргументом**. Параметр — пустое поле на бланке, аргумент — то, что вписали.

## 3. `print` и `return` — главная ловушка урока

- `print` **показывает** значение человеку на экране.
- `return` **отдаёт** значение программе: его можно сохранить, сравнить, передать в другую функцию.

> **Аналогия.** `print` — крикнуть «суп готов!». `return` — передать тарелку следующему повару. Если только крикнуть, следующий повар ничего не получит.

````python
def average_print(values):
    print(sum(values) / len(values))      # только показывает

def average_return(values):
    return sum(values) / len(values)      # отдаёт результат

scores = [4, 5, 3, 5, 4]

a = average_print(scores)
print("Функция с print вернула:", a)

b = average_return(scores)
print("Функция с return вернула:", b)
print("Можно округлить:", round(b, 1))
print("Можно сравнить:", b >= 4)
````

Функция без `return` возвращает специальное значение **`None`** — «ничего». С `None` нельзя считать: `None + 1` даст ошибку. Если результат функции нужен дальше — нужен `return`.

## 4. Несколько параметров и значения по умолчанию

У функции может быть несколько параметров. Если параметр почти всегда одинаковый, ему дают **значение по умолчанию**:

````python
def is_pass(score, threshold=60):
    return score >= threshold

print(is_pass(75))                 # threshold = 60 по умолчанию
print(is_pass(75, 80))             # передали свой порог
print(is_pass(75, threshold=80))   # то же, но с именем — понятнее
````

`return score >= threshold` сразу возвращает результат сравнения — `True` или `False` (урок 4). Писать `if score >= threshold: return True else: return False` не нужно: это то же самое, только длиннее.

Параметры **без** значения по умолчанию пишутся **первыми**.

## 5. Несколько `return` и ранний выход

`return` сразу заканчивает функцию. Этим удобно обрабатывать особые случаи **в начале**:

````python
def average(values):
    if len(values) == 0:
        return None                # особый случай: данных нет
    return sum(values) / len(values)

print(average([10, 20, 30]))
print(average([]))
````

````python
def grade_label(score):
    if score < 0 or score > 100:
        return "ошибка в данных"
    if score >= 90:
        return "отлично"
    if score >= 60:
        return "зачёт"
    return "нужно повторить"

for s in [95, 75, 40, 120]:
    print(s, "→", grade_label(s))
````

## 6. Область видимости

Переменные, созданные **внутри** функции, — **локальные**: снаружи их не видно. Функция общается с внешним миром только через **параметры** (вход) и **`return`** (выход).

````python
def make_label(score):
    if score >= 60:
        status = "зачёт"
    else:
        status = "незачёт"
    return f"{score}: {status}"

label = make_label(55)
print(label)
# print(status)   # раскомментируй: NameError — status живёт только внутри функции
````

Зачем так сделано: функции не мешают друг другу, даже если внутри у них переменные с одинаковыми именами. Это упрощает поиск ошибок.

## 7. Описание функции

Хорошая привычка — первой строкой функции коротко написать, что она делает. Это **строка документации**, она пишется в тройных кавычках. `help(имя_функции)` покажет её.

````python
def normalize(text):
    """Убрать пробелы по краям, перевести в нижний регистр, убрать ! и ."""
    text = text.strip().lower()
    text = text.replace("!", "").replace(".", "")
    return text

print(normalize("  Да! "))
help(normalize)
````

## 8. Конвейер из функций

Большую задачу делим на маленькие шаги, каждый шаг — функция с понятным именем:

1. `parse_numbers` — разобрать строку в список чисел;
2. `clean_scores` — оставить только допустимые значения;
3. `average` — посчитать итог.

Каждую функцию можно проверить отдельно, а сломавшуюся — заменить, не трогая остальные. Такую цепочку называют **конвейером** (по-английски *pipeline*).

````python
def parse_numbers(text):
    """'5 4 9' -> [5, 4, 9]"""
    numbers = []
    for part in text.split():
        numbers.append(int(part))
    return numbers

def clean_scores(values, low=1, high=5):
    """Оставить только оценки от low до high."""
    result = []
    for v in values:
        if low <= v <= high:
            result.append(v)
    return result

raw = "5 4 9 3 5 -1 4 2"
parsed = parse_numbers(raw)
cleaned = clean_scores(parsed)
print("Разобрали:", parsed)
print("Почистили:", cleaned)
print("Среднее:", round(average(cleaned), 2))
````

Запись `low <= v <= high` — двойное сравнение: «v между low и high». Это то же самое, что `low <= v and v <= high`.

## 9. Тесты для функции

Функцию удобно проверять набором тестов: список входов и список ожидаемых ответов.

````python
inputs = [95, 90, 89, 60, 59, -5]
expected = ["отлично", "отлично", "зачёт", "зачёт", "нужно повторить", "ошибка в данных"]

passed = 0
for i, value in enumerate(inputs):
    got = grade_label(value)
    if got == expected[i]:
        passed += 1
        print("OK  ", value, "→", got)
    else:
        print("FAIL", value, "→", got, "| ожидали:", expected[i])
print(f"Пройдено тестов: {passed} из {len(inputs)}")
````

## 10. Хорошие имена функций

Имя функции — это **действие**: по нему должно быть понятно, что она делает.

| Плохо | Хорошо |
|-------|--------|
| `f1`, `do`, `stuff` | `clean_text`, `count_errors` |
| `func(a, b)` | `safe_divide(numerator, denominator)` |
| `check(x)` | `is_pass(score)`, `needs_human(errors)` |

Функции, которые возвращают `True` / `False`, часто называют с `is_`, `has_`, `needs_`.

## 11. Частые ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|---------------|
| Забыли `return` | функция возвращает `None` | добавить `return` |
| `print` вместо `return` | дальше в программе `None` | `print` — для человека, `return` — для программы |
| Объявили, но не вызвали | ничего не происходит | вызвать: `имя(аргументы)` |
| Вызвали до запуска ячейки с `def` | `NameError` | сначала запустить ячейку с объявлением |
| Используем внутреннюю переменную снаружи | `NameError` | вернуть её через `return` |
| Одна функция делает всё | трудно проверить | разбить на шаги |
