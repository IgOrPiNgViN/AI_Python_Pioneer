## 1. Словарь — карточка с полями

Сравни два способа хранить данные об ученике:

```python
student_list = ["Ира", 14, 88]                         # что на каком месте?
student = {"name": "Ира", "age": 14, "score": 88}      # всё подписано
```

**Словарь** (`dict`) хранит пары **ключ → значение**. Ключ — подпись поля (обычно строка), значение — что угодно: число, строка, список.

| Действие | Код |
|----------|-----|
| создать | `d = {"name": "Ира", "score": 88}` |
| прочитать | `d["name"]` |
| изменить или добавить | `d["score"] = 90`, `d["city"] = "Москва"` |
| безопасно прочитать | `d.get("age", "нет данных")` |
| есть ли ключ | `"age" in d` |
| сколько пар | `len(d)` |

````python
student = {"name": "Ира", "age": 14, "score": 88}
print(student["name"])
print(student["score"])

student["score"] = 90           # изменить значение
student["city"] = "Москва"      # добавить новое поле
print(student)

print(student.get("phone", "нет данных"))   # такого ключа нет — вернётся запасное значение
print("phone" in student)
````

Если обратиться к несуществующему ключу через квадратные скобки — будет ошибка `KeyError`. Чаще всего это **опечатка** в имени поля: `"Score"` вместо `"score"`.

Метод `keys()` выдаёт все ключи словаря в особом «наборе». Функция `list(...)` превращает такой набор в **обычный список**, чтобы его было удобно напечатать. `list` пригодится ещё не раз: она делает список из всего, что можно перебрать циклом.

````python
# Раскомментируй, прочитай ошибку и верни # на место:
# print(student["Score"])
print("Ключи:", list(student.keys()))
````

### Перебор словаря

````python
for key in student:
    print(key, "→", student[key])

print("---")
for key, value in student.items():      # сразу пары ключ, значение
    print(f"{key}: {value}")
````

Обычный цикл по словарю перебирает **ключи**, а значение достаём по ключу: `student[key]`. Метод `items()` выдаёт **пары** — поэтому, как с `enumerate` на уроке 6, в цикле две переменные.

## 2. Словарь-счётчик

Очень частая задача: **сколько раз встречается каждое значение**. Какие слова чаще в отзывах? Сколько учеников из каждого города? Решение — словарь, где ключ — значение, а число — сколько раз встретили.

````python
text = "данные модель данные проверка модель данные"
counts = {}
for word in text.split():
    if word in counts:
        counts[word] += 1       # уже встречали — увеличиваем
    else:
        counts[word] = 1        # встретили впервые
print(counts)
````

Этот приём — основа анализа текста: позже на курсе мы научим модель работать с текстом, превращая его именно в такие счётчики слов.

## 3. Список словарей — таблица в памяти

Если у каждого ученика есть карточка-словарь, весь класс — это **список словарей**. Каждый словарь — строка таблицы, ключи — названия столбцов.

| name | score | city |
|------|------:|------|
| Ира | 88 | Москва |
| Олег | 74 | Казань |
| Аня | 91 | Москва |

````python
students = [
    {"name": "Ира", "score": 88, "city": "Москва"},
    {"name": "Олег", "score": 74, "city": "Казань"},
    {"name": "Аня", "score": 91, "city": "Москва"},
]

for s in students:                      # s — по очереди каждый словарь-строка
    print(s["name"], s["score"], s["city"])

total = 0
for s in students:
    total += s["score"]
print("Средний балл:", round(total / len(students), 1))

moscow = []
for s in students:
    if s["city"] == "Москва":
        moscow.append(s["name"])
print("Из Москвы:", moscow)
````

## 4. Файлы: запись и чтение

```python
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("Первая строка\n")
```

- `open(имя, режим, encoding="utf-8")` открывает файл;
- `as f` — даём открытому файлу короткое имя `f`, через него пишем `f.write(...)` и читаем `f.read()`;
- режимы: `"r"` — читать (по умолчанию), `"w"` — записать **заново** (старое содержимое стирается!), `"a"` — дописать в конец;
- `encoding="utf-8"` — чтобы русские буквы не превратились в «кракозябры»;
- `with` сам **закроет** файл, когда блок закончится;
- `"\n"` — символ переноса строки.

````python
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("Первая строка\n")
    f.write("Вторая строка\n")
    f.write("Курс: Инженерия ИИ\n")

with open("notes.txt", encoding="utf-8") as f:
    content = f.read()
print(content)
````

````python
# Файл можно читать построчно в цикле
with open("notes.txt", encoding="utf-8") as f:
    for line in f:
        print(repr(line))
````

Видишь `\n` в конце каждой строки? Это перенос строки, который мы записали. Его убирают знакомым методом `strip()`.

## 5. CSV — таблица в текстовом файле

**CSV** (*comma-separated values*) — «значения через запятую». Первая строка — названия столбцов, дальше — строки данных:

```text
name,score,city
Ира,88,Москва
Олег,74,Казань
```

CSV открывается в Excel и Google Таблицах, его легко читать из Python. Это самый частый формат учебных данных.

Сначала создадим файл — запиши его построчно:

````python
with open("students.csv", "w", encoding="utf-8") as f:
    f.write("name,score,city\n")
    f.write("Ира,88,Москва\n")
    f.write("Олег,74,Казань\n")
    f.write("Аня,91,Москва\n")
    f.write("Боря,55,Тула\n")
    f.write("Маша,67,Казань\n")
    f.write("Кира,82,Тула\n")
    f.write("Никита,49,Москва\n")
print("Файл students.csv создан")
````

### Чтение CSV модулем `csv`

Можно разбирать каждую строку через `split(",")`, но в Python есть готовый модуль `csv` (подключаем его через `import`, как `random` на уроке 5). `csv.DictReader` превращает каждую строку файла в **словарь**, а ключи берёт из заголовка.

````python
import csv

with open("students.csv", encoding="utf-8", newline="") as f:
    rows = list(csv.DictReader(f))

print("Строк:", len(rows))
print("Первая строка:", rows[0])
print("Тип балла:", type(rows[0]["score"]))
````

Разберём:

- `csv.DictReader(f)` — «читатель», который выдаёт строки файла по одной в виде словарей;
- `list(...)` собирает их все в список — получаем таблицу в памяти, как в разделе 3;
- `rows[0]` — первая строка (словарь), а `rows[0]["score"]` — значение столбца `score` в ней: сначала индекс строки, потом ключ столбца.

**Ловушка, как в уроке 3:** все значения из CSV — **строки**. `"88"`, а не `88`. Перед расчётами преобразуем числа — сразу после чтения:

````python
for row in rows:
    row["score"] = int(row["score"])

print(rows[0], type(rows[0]["score"]))
````

`newline=""` — рекомендация для модуля `csv`: так строки файла читаются правильно в любой системе.

## 6. Считаем по таблице

После чтения из таблицы извлекают смысл: среднее, долю, фильтр, группировку.

````python
# Среднее и максимум
total = 0
best = rows[0]
for row in rows:
    total += row["score"]
    if row["score"] > best["score"]:
        best = row
print("Средний балл:", round(total / len(rows), 1))
print("Лучший:", best["name"], best["score"])

# Новое поле status и доля зачётов
passed = 0
for row in rows:
    if row["score"] >= 60:
        row["status"] = "зачёт"
        passed += 1
    else:
        row["status"] = "пересдача"
print("Доля зачётов:", round(passed / len(rows), 2))
````

### Группировка: средний балл по городам

Словарь, где ключ — город, а значение — **список** баллов учеников этого города. Когда город встречается впервые, кладём для него пустой список, а потом добавляем в этот список баллы через `append`:

````python
by_city = {}
for row in rows:
    city = row["city"]
    if city not in by_city:
        by_city[city] = []
    by_city[city].append(row["score"])

for city, scores in by_city.items():
    print(f"{city}: учеников {len(scores)}, средний балл {round(sum(scores) / len(scores), 1)}")
````

## 7. Сохраняем отчёт в новый CSV

Частый цикл работы с данными: **прочитал → посчитал → сохранил результат**. `csv.DictWriter` записывает список словарей в файл:

````python
with open("report.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score", "status"], extrasaction="ignore")
    writer.writeheader()          # строка заголовков
    writer.writerows(rows)        # все строки

with open("report.csv", encoding="utf-8") as f:
    print(f.read())
````

- `fieldnames` — список столбцов, которые попадут в файл, в нужном порядке;
- `writeheader()` пишет первую строку с названиями столбцов;
- `writerows(rows)` пишет все словари из списка — по строке на словарь;
- `extrasaction="ignore"` значит: если в словаре есть лишние поля (у нас — `city`), их не записывать.

## 8. Всё вместе: функции для работы с таблицей

Соберём то, что умеем, в конвейер из функций — так будет удобно на СР №1.

````python
def load_students(path):
    """Прочитать CSV и превратить score в число."""
    with open(path, encoding="utf-8", newline="") as f:
        rows = list(csv.DictReader(f))
    for row in rows:
        row["score"] = int(row["score"])
    return rows

def summary(rows, threshold=60):
    """Вернуть словарь-сводку по таблице."""
    if len(rows) == 0:
        return {"count": 0, "average": None, "pass_rate": 0}
    total = 0
    passed = 0
    for row in rows:
        total += row["score"]
        if row["score"] >= threshold:
            passed += 1
    return {
        "count": len(rows),
        "average": round(total / len(rows), 1),
        "pass_rate": round(passed / len(rows), 2),
    }

data = load_students("students.csv")
result = summary(data)
print(result)
print("Средний балл:", result["average"])
````

Функция возвращает **словарь** — так удобно отдавать сразу несколько результатов, и у каждого есть понятное имя.

## 9. Частые ошибки

| Открыли нужный файл в режиме `"w"` | файл стёрт | для чтения — режим по умолчанию |
| Нет `encoding="utf-8"` | «кракозябры» вместо русских букв | всегда указывать кодировку |
