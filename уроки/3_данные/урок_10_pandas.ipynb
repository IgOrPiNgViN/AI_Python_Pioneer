## 3. CSV: сохранить и прочитать

На уроке 8 мы писали CSV вручную построчно. В pandas — одной командой:

- `df.to_csv("файл.csv", index=False)` — сохранить (`index=False` — не записывать номера строк отдельным столбцом);
- `pd.read_csv("файл.csv")` — прочитать.

`read_csv` сам определяет, где числа, а где текст: преобразовывать типы вручную, как на уроке 8, обычно не нужно.

````python
df.to_csv("journal.csv", index=False)

df = pd.read_csv("journal.csv")
print("Прочитали строк:", len(df))
df.head(3)
````

## 4. Паспорт данных — первые команды с любой таблицей

Прежде чем что-то считать, инженер **осматривает** данные. Как врач: сначала температура и пульс, потом лечение.

| Команда | На какой вопрос отвечает |
|---------|--------------------------|
| `df.shape` | сколько строк и столбцов? |
| `df.columns` | как называются столбцы? нет ли опечаток? |
| `df.head()` / `df.tail()` | как выглядят первые / последние строки? |
| `df.info()` | какие типы у столбцов? есть ли пустые ячейки? |
| `df.describe()` | разумны ли числа: минимум, максимум, среднее? |

Обрати внимание: `df.shape` и `df.columns` пишутся **без скобок** — это **свойства** таблицы, готовые сведения о ней. А `df.head()`, `df.info()` — **методы** (урок 6), они что-то делают, поэтому со скобками. В скобках `head(3)` можно указать, сколько строк показать.

````python
print("Размер (строки, столбцы):", df.shape)
print("Столбцы:", list(df.columns))
````

````python
df.tail(3)
````

````python
df.info()
````

### Как читать `info()`

- **Non-Null Count** — сколько заполненных ячеек. Если меньше числа строк — есть **пропуски**. Сегодня таблица чистая; на уроке 11 будет «грязная».
- **Dtype** — тип столбца: `int64` — целые числа, `float64` — дробные, `object` — текст (в новых версиях pandas текст обозначается `str`).

````python
df.describe()
````

### Как читать `describe()`

| Строка | Смысл |
|--------|-------|
| `count` | сколько значений |
| `mean` | среднее |
| `std` | разброс: насколько значения отличаются от среднего |
| `min` / `max` | минимум и максимум — первое место, где видны ошибки и **выбросы** (значения, резко непохожие на остальные) |
| `25%` / `50%` / `75%` | четверти: `50%` — это **медиана**, «серединное» значение |

Вопросы инженера к `describe()`: нет ли балла 150 при шкале 0–100? Нет ли отрицательных часов? Похоже ли среднее на «типичного» ученика?

## 5. Столбцы и статистики

`df["балл"]` — один столбец (Series). У него есть готовые функции: `mean()`, `median()`, `min()`, `max()`, `sum()`.

````python
scores = df["балл"]                      # один столбец — Series
print("Среднее:", round(scores.mean(), 1))
print("Медиана:", scores.median())
print("Минимум и максимум:", scores.min(), scores.max())
print("Всего часов у всех:", df["часы"].sum())
````

### Среднее или медиана?

Медиана — значение «посередине», если выстроить всех по порядку. Она **устойчива к выбросам**. Если у одного ученика по ошибке записали балл 900, среднее сильно вырастет, а медиана почти не изменится. Это пригодится на уроке 11, когда будем чинить данные.

### Несколько столбцов

Чтобы взять несколько столбцов, передаём **список** их названий — поэтому скобки двойные:

````python
df[["имя", "кружок", "балл"]].head()
````

**Частая ошибка:** `df["имя", "балл"]` — без внутренних скобок. Будет `KeyError`. Правильно: `df[["имя", "балл"]]`.

## 6. Фильтры: выбираем строки по условию

Условие над столбцом даёт **маску** — столбец из `True` и `False` для каждой строки. Таблица с маской в скобках оставляет только строки с `True`.

````python
mask = df["балл"] >= 80
print(mask.head())
print("Сколько True:", mask.sum())
````

`mask.sum()` считает, сколько в маске `True`: при сложении Python считает `True` за 1, а `False` за 0.

````python
df[df["балл"] >= 80]
````

### Несколько условий

Для таблиц вместо `and`, `or`, `not` пишут значки:

| Обычный Python | pandas | Смысл |
|:--------------:|:------:|-------|
| `and` | `&` | и |
| `or` | `\|` | или |
| `not` | `~` | не |

Каждое условие — **в своих скобках**, иначе будет ошибка.

````python
# Девятиклассники из кружка Python
df[(df["класс"] == 9) & (df["кружок"] == "Python")]
````

````python
# Мало часов ИЛИ низкая посещаемость — кандидаты на помощь
df[(df["часы"] <= 3) | (df["посещаемость"] < 60)]
````

````python
# isin — значение из списка
df[df["кружок"].isin(["Роботы", "Игры"])].head()
````

## 7. Сортировка

`sort_values("столбец")` сортирует по возрастанию, `ascending=False` — по убыванию. Вместе с `head()` получаем «топ».

````python
df.sort_values("балл", ascending=False).head(5)
````

Сортировка **не меняет** `df` — она возвращает новую таблицу. Чтобы сохранить результат, его присваивают: `df_sorted = df.sort_values("балл")`. Так же ведут себя и фильтры.

## 8. Частоты категорий: `value_counts`

Для текстовых столбцов первая команда — «каких значений сколько?». `normalize=True` даёт доли вместо количеств.

````python
print(df["кружок"].value_counts())
print()
print(df["кружок"].value_counts(normalize=True))
````

Это то же самое, что словарь-счётчик из урока 8, — только одной строкой.

## 9. Новые столбцы

Новый столбец создаётся присваиванием, как новый ключ в словаре. Значение можно вычислить из других столбцов — сразу для всех строк, без цикла.

Метод `.round(1)` у столбца округляет все значения сразу — как `round()` из урока 3.

````python
df["сдал"] = df["балл"] >= 60
df["балл_за_час"] = (df["балл"] / df["часы"]).round(1)
df[["имя", "балл", "часы", "сдал", "балл_за_час"]].head(6)
````

````python
print(df["сдал"].value_counts())

# Сохраним только тех, кто сдал: столбец "сдал" уже состоит из True / False,
# поэтому сам является маской
passed = df[df["сдал"]]
passed.to_csv("passed.csv", index=False)
print("Сохранено строк:", len(passed))
````

## 10. Частые ошибки

| Ошибка | Что происходит | Как исправить |
|--------|----------------|---------------|
| `df["имя", "балл"]` | `KeyError` | `df[["имя", "балл"]]` |
| `and` / `or` в фильтре | `ValueError` | `&` / `\|` и скобки вокруг каждого условия |
| Опечатка в названии столбца | `KeyError` | сверить с `df.columns` |
| `df[df["кружок"] == Python]` | `NameError` | текст — в кавычках: `"Python"` |
| Ждали, что `sort_values` изменит `df` | порядок не изменился | сохранить результат в переменную |
