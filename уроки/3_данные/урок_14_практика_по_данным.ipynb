<details>
<summary>Пример решения</summary>

```python
clean = df.copy()
clean["спорт"] = clean["спорт"].str.strip().str.lower()
print(clean["спорт"].value_counts())

clean["экран"] = clean["экран"].replace({"четыре": "4"})
clean["экран"] = pd.to_numeric(clean["экран"], errors="coerce")

before = len(clean)
clean = clean.drop_duplicates(subset=["id"], keep="first")
print("Удалено дубликатов:", before - len(clean))

print(clean[(clean["сон"] < 3) | (clean["сон"] > 14)])
clean = clean[(clean["сон"] >= 3) & (clean["сон"] <= 14)]

clean["экран"] = clean["экран"].fillna(clean["экран"].median())
clean = clean.dropna(subset=["настроение"])
clean = clean.reset_index(drop=True)

print(clean.shape)
print(clean.isna().sum())
clean.to_csv("survey_clean.csv", index=False)
```

| Проблема | Что сделал | Сколько строк | Почему |
|----------|------------|--------------:|--------|
| «Да», «да », «НЕТ» | `strip` + `lower` | 4 | одно значение — одна запись |
| «четыре» в `экран` | заменил на 4 | 1 | смысл очевиден |
| анкета `id = 9` дважды | удалил повтор | 1 | один человек |
| сон 75 часов | удалил строку | 1 | невозможно; 7.5 или 7? правды не знаем |
| пропуск в `экран` | медиана | 1 | одно значение, медиана устойчива |
| пропуск в `настроение` | удалил строку | 1 | это главный показатель опроса, выдумывать нельзя |

````python
# Станция 3
````
