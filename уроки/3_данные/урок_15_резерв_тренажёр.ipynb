## Часть B. Задачник по pandas (уроки 10–13)

Таблица — «Журнал кружков» из урока 10.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "имя": ["Аня", "Боря", "Вика", "Глеб", "Даша", "Егор", "Женя", "Зоя", "Илья", "Кира",
            "Лёша", "Мила", "Никита", "Оля", "Петя", "Рита", "Серёжа", "Таня", "Уля", "Фёдор"],
    "класс": [8, 9, 8, 9, 8, 8, 9, 9, 8, 9, 8, 9, 9, 8, 8, 9, 8, 9, 9, 8],
    "кружок": ["Python", "Роботы", "Python", "Игры", "Роботы", "Игры", "Python", "Игры", "Роботы", "Python",
               "Python", "Игры", "Роботы", "Игры", "Python", "Роботы", "Игры", "Python", "Роботы", "Игры"],
    "часы": [8, 3, 10, 5, 7, 2, 9, 6, 4, 11, 8, 5, 12, 3, 7, 9, 4, 10, 6, 5],
    "посещаемость": [90, 55, 95, 70, 80, 40, 88, 75, 60, 98, 85, 72, 100, 50, 78, 92, 65, 94, 70, 68],
    "балл": [85, 42, 91, 58, 74, 35, 88, 69, 51, 96, 82, 63, 98, 40, 71, 90, 55, 93, 67, 60],
})
df.head()
````

Каждую задачу реши в отдельной ячейке (добавь её кнопкой «+ Код»).

1. Сколько учеников в 9 классе?
2. Средний балл учеников, которые занимаются меньше 5 часов в неделю.
3. Имена трёх учеников с самой низкой посещаемостью.
4. Сколько учеников набрали от 60 до 80 баллов включительно?
5. Добавь столбец `часы_в_месяц` (в месяце примерно 4 недели).
6. Какой кружок самый большой? (`value_counts`)
7. Средний балл и число учеников по классам (`groupby` + `agg`).
8. Корреляция посещаемости и балла.
9. Гистограмма часов с подписями.
10. Есть ли ученики с посещаемостью ниже 60, но баллом выше 70? Что бы это могло значить?

````python
# Задачи B1–B10
````

<details>
<summary>Решения части B</summary>

```python
print("1:", (df["класс"] == 9).sum())
print("2:", round(df[df["часы"] < 5]["балл"].mean(), 1))
print("3:", df.sort_values("посещаемость")["имя"].head(3).tolist())
print("4:", len(df[(df["балл"] >= 60) & (df["балл"] <= 80)]))
df["часы_в_месяц"] = df["часы"] * 4
print("6:", df["кружок"].value_counts())
print("7:", df.groupby("класс")["балл"].agg(["mean", "count"]).round(1))
print("8:", round(df["посещаемость"].corr(df["балл"]), 2))

plt.figure(figsize=(6, 4))
plt.hist(df["часы"], bins=6, edgecolor="black")
plt.title("Часы занятий в неделю")
plt.xlabel("Часов в неделю")
plt.ylabel("Число учеников")
plt.show()

print("10:", df[(df["посещаемость"] < 60) & (df["балл"] > 70)])
```
В задаче 10 таких учеников нет. Если бы нашлись — это повод проверить данные: возможно, ошибка ввода, а возможно, ученик занимается сам дома.

## Часть C. Как спросить ИИ-помощника об ошибке в данных

На уроке 2 мы учили схему запроса: **роль → контекст → задача → формат → ограничения**. С данными особенно важен **контекст**: помощник не видит твою таблицу и легко «придумает» столбцы, которых нет.

Что приложить к вопросу:

- названия столбцов: `print(df.columns.tolist())`;
- типы: `print(df.dtypes)`;
- 2–3 первые строки **без личных данных**;
- **точный** текст ошибки и строку кода, где она возникла.
