# Урок 30. Итоговый проект: выбор темы и паспорт проекта

**Дата:** 15.04.2027 · **60 минут** · 8–9 класс  
**Модуль 6.** Итоговый проект (уроки 30–36)  
**Опирается на:** весь курс; особенно урок 1 (цикл инженера), урок 17 (паспорт задачи), урок 29 (работа с ИИ).

Итоговый проект — это **весь цикл инженера ИИ** на твоей собственной задаче: задача → данные → модель → проверка → демонстрация → ограничения. 27 мая ты защитишь его перед классом. Сегодня — самый важный шаг: выбрать **реалистичную** тему и заполнить **паспорт проекта**.

## Цели урока

1. Понять, что требуется от проекта и как он будет оцениваться.
2. Выбрать тип проекта и тему из банка или придумать свою.
3. Сузить тему до задачи, которую можно сделать за 4 занятия.
4. Заполнить паспорт проекта и получить одобрение преподавателя.

## 1. Каким должен быть проект

**Хороший проект 8–9 класса** — не самый сложный, а **доведённый до конца и честный**:

- понятная задача и пользователь;
- реальные или честно собранные данные (хотя бы 50–100 примеров для таблиц, 20–40 на класс для картинок и текстов);
- модель сравнена с baseline;
- метрики объяснены словами;
- работающее демо на 2–3 минуты;
- названы ограничения и риски.

**Плохой проект** — «сделаю ИИ, который понимает всё», данные из ниоткуда, нет проверки, 100% accuracy без объяснения (а значит, скорее всего, утечка).

Можно работать **один** или **в паре** (тогда каждый должен понимать весь проект и отвечать на вопросы по любой части).

## 4. Типы проектов

| Тип | Что делаешь | Инструменты из курса | Минимум данных |
|-----|-------------|----------------------|----------------|
| **Таблица → классификация** | предсказать класс (да / нет, категория) | pandas, дерево, kNN, метрики классификации (уроки 10–21) | 80–100 строк |
| **Таблица → регрессия** | предсказать число | pandas, линейная регрессия, MAE (урок 22) | 80–100 строк |
| **Картинки** | распознать класс на изображении | Teachable Machine или numpy + kNN (урок 26) | 30–50 фото на класс |
| **Текст / бот** | понять намерение и ответить | правила + мешок слов + kNN (урок 27) | 3–5 намерений по 10–15 фраз |

Любой тип проходит **одинаковый цикл** и оценивается по тем же критериям.

## 5. Банк тем

**Таблицы:**
- Предсказать, успеет ли ученик сдать проект в срок, по анкете о нагрузке (классификация).
- Предсказать время на домашнее задание по предмету, объёму и усталости (регрессия).
- Предсказать, понравится ли фильм / книга / игра, по жанру, длительности, рейтингу (классификация).
- Предсказать, сколько минут займёт дорога до школы, по погоде, транспорту и времени выхода (регрессия).
- Учебный датасет из sklearn (например, «Ирисы» `load_iris` или «Вина» `load_wine`) с подробной разведкой и честным сравнением моделей — подходит, если не получается собрать свои данные.

**Картинки:**
- Жесты руки для управления презентацией (Teachable Machine).
- Сортировка мусора: пластик / бумага / металл по фото предметов.
- Распознавание собственных рукописных цифр или букв.

**Текст:**
- Бот-помощник для кружка, библиотеки, спортивной секции, столовой.
- Определение тональности отзывов о школьной столовой.
- Классификатор вопросов к учителю по темам курса.

Можно придумать **свою** тему — главное, чтобы она прошла проверку из раздела 6.

## 6. Сужаем тему

Большинство неудачных проектов проваливаются из-за слишком широкой темы. Пять вопросов-фильтров:

1. **Кому** это поможет и **в какой момент** он будет пользоваться моделью?
2. **Что ровно** предсказываем? Одна цель y.
3. **Где возьму данные** за неделю? (опрос одноклассников, свои фото, открытый датасет, учебный датасет)
4. **Как пойму, что получилось?** Какая метрика и с чем сравниваю (baseline)?
5. **Успею ли** за 4 занятия и 4 вечера?

| Широко (плохо) | Узко (хорошо) |
|----------------|---------------|
| «ИИ для школы» | «Бот, который отвечает на 5 типовых вопросов о расписании кружков» |
| «Предсказывать оценки» | «Предсказать, напишет ли ученик пробник по математике на 4+, по часам подготовки и числу решённых вариантов» |
| «Распознавать всё на фото» | «Отличать 3 жеста руки при свете в классе» |

### Антипаттерны

- данные, которые нельзя собрать за неделю;
- 20 классов вместо 3;
- задача, где модели нечего «выучить» (например, предсказать бросок кубика);
- данные с фото лиц и личной информацией без согласия.

## 8. Паспорт проекта

Паспорт — главный документ проекта. По нему преподаватель одобряет тему, а жюри на защите понимает, что ты делал.

**Пример заполненного паспорта:**

| Пункт | Пример |
|-------|--------|
| Название | «Успею ли я?» — прогноз сдачи проекта в срок |
| Пользователь | ученик 8–9 класса, который планирует неделю |
| Проблема | ученики не рассчитывают время и сдают проекты с опозданием |
| Момент предсказания | за неделю до дедлайна |
| Цель y | успеет в срок (да / нет) |
| Тип задачи | бинарная классификация |
| Признаки X | дней до дедлайна, сколько уже сделано (%), уроков в день, кружков в неделю |
| Не берём | фактическую дату сдачи (утечка), ФИО (персональные данные) |
| Данные | анонимный опрос в Google Форме, 60–100 ответов, с согласием |
| Baseline | «всегда успеет» |
| Главная метрика | recall для класса «не успеет»: важнее предупредить |
| Критерий успеха | recall не ниже 0.7 и лучше baseline |
| Демо | ввожу свои данные — модель отвечает и объясняет, на что обратить внимание |
| Риски | мало данных; ответы в опросе неточные; нельзя использовать для оценок |
| Работа с ИИ | объяснение ошибок, идеи признаков; веду журнал использования ИИ |

## 9. Чек-лист одобрения темы

Преподаватель одобряет тему, если:

- [ ] есть конкретный пользователь и момент предсказания;
- [ ] одна цель y и понятный тип задачи;
- [ ] данные реально собрать или взять до 22.04;
- [ ] нет утечек в признаках;
- [ ] нет сбора лишних персональных данных; есть план получения согласия;
- [ ] есть baseline и главная метрика;
- [ ] объём работы реалистичен.
