## 1. Демо

Хорошее демо — это **короткий сценарий**, а не «вот мой код»:

1. **Кто пользователь и какая у него проблема** (1 фраза).
2. **Ввод** — показываем 1–2 типичных примера: модель отвечает правильно.
3. **Сложный случай** — честно показываем, где модель ошибается.
4. **Вывод** — что модель умеет и где ей нельзя доверять.

### Шаблон интерактивного демо (табличный проект)

Ниже — пример: модель спрашивает пользователя о его ситуации и отвечает. Замени обучение на свою модель из урока 32 (или скопируй ячейки обучения в этот ноутбук).

````python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Пример обучающих данных — замени на свою модель из урока 32
train = pd.DataFrame({
    "дней_до_дедлайна": [1, 2, 3, 4, 6, 8, 9, 10, 12, 14, 3, 7],
    "сделано_процентов": [10, 80, 30, 20, 60, 40, 30, 50, 10, 70, 90, 15],
    "успел": [False, True, False, False, True, True, True, True, True, True, True, False],
})
model = DecisionTreeClassifier(max_depth=2, random_state=0)
model.fit(train[["дней_до_дедлайна", "сделано_процентов"]], train["успел"])

def demo():
    """Спросить пользователя и показать прогноз модели."""
    print("Прогноз: успеешь ли ты сдать проект в срок?")
    days = int(input("Сколько дней до дедлайна? "))
    done = int(input("Сколько процентов работы уже сделано? "))
    if days < 1 or days > 30 or done < 0 or done > 100:
        print("Модель обучена на 1–30 днях и 0–100 процентах. Для таких значений прогноз ненадёжен.")
        return
    person = pd.DataFrame({"дней_до_дедлайна": [days], "сделано_процентов": [done]})
    if model.predict(person)[0]:
        print("Скорее всего успеешь. Но проверь план на каждый день!")
    else:
        print("Есть риск не успеть. Составь план и начни с самого большого куска работы.")
    print("Помни: это прогноз учебной модели на маленьких данных, а не приговор.")

demo()
````

Обрати внимание на детали, которые делают демо **инженерным**:

- проверка ввода: модель не отвечает на значения вне диапазона обучающих данных (урок 22 — экстраполяция);
- ответ сформулирован как **совет**, а не приговор;
- в конце — честное предупреждение об ограничениях.

**Для бота:** демо — это диалог из 5–6 реплик, включая вопрос не по теме (бот должен ответить «не понял»).  
**Для картинок:** демо — показ 2–3 предметов в камеру, включая один сложный случай (другой фон, необычный ракурс).
