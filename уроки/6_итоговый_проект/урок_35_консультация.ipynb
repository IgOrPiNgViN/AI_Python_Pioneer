## 1. Итоговый чек-лист проекта

Отметь честно. Каждый пустой пункт — задача на неделю.

**Постановка**
- [ ] есть паспорт: пользователь, проблема, момент предсказания, цель y, тип задачи;
- [ ] критерий успеха записан **до** обучения модели.

**Данные**
- [ ] журнал данных заполнен: источник, размер до / после, очистка, правила разметки;
- [ ] нет утечек; исключённые столбцы объяснены;
- [ ] данные обезличены, есть согласие участников;
- [ ] есть хотя бы один график разведки с выводом.

**Модель / бот / распознавание**
- [ ] есть baseline и его метрика;
- [ ] модель проверена на test (или на отдельных фото / фразах);
- [ ] настройки выбраны честно (не подбирались по test);
- [ ] сравнены train и test; известно, есть ли переобучение;
- [ ] разобраны ошибки: где и почему модель ошибается.

**Демо и выступление**
- [ ] демо работает вживую и проходит 5 тестов «сломай своё демо»;
- [ ] выступление укладывается в 5 минут;
- [ ] 4–5 слайдов с крупным текстом.

**Документы**
- [ ] карточка метрик и карточка ограничений;
- [ ] журнал модели;
- [ ] журнал использования ИИ;
- [ ] ноутбук проходит «Перезапустить сеанс и выполнить все» без ошибок.

**Понимание**
- [ ] могу объяснить каждую ячейку своего кода;
- [ ] могу объяснить метрику человеку без подготовки.

## 2. Если времени мало: приоритеты

Сначала то, что даёт больше всего баллов и без чего проект «не стоит»:

1. **Работающий ноутбук** сверху вниз и **демо**.
2. **Baseline и метрики** на test, объяснённые словами.
3. **Карточка ограничений** с конкретными пунктами.
4. Журналы данных и модели.
5. Красивые слайды — в последнюю очередь.

**Чего не делать за неделю до защиты:** менять тему, переписывать всё «на более крутую модель», собирать новые данные с нуля. Лучше честно описать ограничения текущей версии.

## 4. Типичные вопросы жюри

Потренируйтесь в парах: один задаёт вопросы, второй отвечает **без подготовки**, потом меняетесь.

| Вопрос | Что хочет услышать жюри |
|--------|-------------------------|
| Кто будет пользоваться вашей моделью? | конкретный человек и ситуация |
| Откуда данные? Сколько? Как вы их проверяли? | источник, размер, очистка, журнал |
| Почему вы не взяли в признаки …? | понимание утечек и приватности |
| Какой у вас baseline? Насколько модель лучше? | конкретные числа |
| Что значит ваша метрика? | объяснение словами в контексте задачи |
| Почему именно эта метрика? | цена ошибки: FN или FP хуже |
| Есть ли переобучение? Как проверяли? | train vs test, validation |
| Где модель ошибается? Почему? | разбор ошибок, примеры |
| Кому может навредить ваша модель? | конкретная группа или ситуация |
| Что сделал ИИ-помощник, а что вы? | журнал использования ИИ |
| Что бы вы улучшили? | конкретный план: данные, признаки, проверка |

### Примеры сильных и слабых ответов

**Вопрос:** «Почему у вас accuracy 0.85 — это хорошо?»

- Слабо: «Ну, 85% — это много».
- Сильно: «Baseline, который всем отвечает „успеет“, даёт 0.6. Наша модель лучше на 25 пунктов. Но для нас важнее recall для „не успеет“ — он 0.79: из 5 учеников с риском модель находит 4».

**Вопрос:** «Можно ли использовать вашу модель в другой школе?»

- Слабо: «Да, наверное».
- Сильно: «Без проверки — нет. Данные собраны в одной школе, 120 человек. В другой школе другая нагрузка, и модель нужно проверить на их данных».
